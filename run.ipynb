#### **Metrics**

- FAR (FPR) - False Alarm Rate (False positive rate) on clean (anomaly-free) data.
- TPR - True Positive Rate (True positive rate) on anomalous data.
- DD - Detection Delay (Time to detect the first anomaly).
- Robustness and resilience:
    - Noise and outliers robustness score (NORS): the relative drop in performance compared to the baseline scenario (e.g., on the clean dataset without noise and/or outliers)
    - Recovery Time (RT): time to recover from an anomaly
    - Stability Time (ST): time until the decision statistic is stable for N consecutive samples

````python
# Read Files by Type
def read_file_auto(fp):
    """Read .csv, .mat, or .txt file and return (filename, DataFrame)
        Runs with the following separators: , ;
    """
    
    fname = os.path.basename(fp)
    if fp.endswith('.csv'):
        df = pd.read_csv(fp, sep = ';' if ';' in open(fp).readline() else ',')
        
    elif fp.endswith('.mat'):
        mat = loadmat(fp)
        data_key = next(k for k in mat.keys() if not k.startswith('__'))
        df = pd.DataFrame(mat[data_key])

    elif fp.endswith('.txt'):
        df = pd.read_csv(fp, header=None, sep='\s+')

    else:
        raise ValueError(f"Unsupported file type: {fp}")
    
    return fname, df
````

````python
# Data Loader Functions
def read_files_threaded(filepaths):
    """ 
        Read files in parallel using ThreadPoolExecutor
        Note: Only supports .csv, .mat, and .txt
    """
    with ThreadPoolExecutor() as executor:
        results = executor.map(read_file_auto, filepaths)

    return dict(results)
````

````python
# Data Loader Standard
def load_data(data_path, dataset=None, multivariate=False):
    """ Load data from file path
        Note: Only supports TEP, LBNL, SKAB, RMFD and VSD (for now)
        To make it easier for you without the need to reorganize the downloaded data.
        You need to: rename the clean files to clean.csv
        Results:
            - For TEP and LBNL:
                - 'clean': A dictionary with one DataFrame (or two TEP)
                - 'fault': A dictionary mapping filenames to DataFrames

            - For RMFD:
                - 'clean': A dictionary containing dictionaries {filename: DataFrame, ...}
                - 'fault': A dictionary containing dictionaries {filename: DataFrame, ...}

            - For SKAB:
                - 'clean': A dictionary with one DataFrame
                - 'fault': A dictionary containing dictionaries, each mapping fault folder name to another dictionary {filename: DataFrame, ...}

            - For VSD:
                - 'clean': A dictionary containing dictionaries {filename: DataFrame, ...}, one per case folder
                - 'fault': A dictionary containing dictionaries {filename: DataFrame, ...}, one per case folder

        File structure (as downloaded from the given sources and very little manipulation):
            data/
                TEP/
                    clean.csv
                    clean_test.csv
                    fault/
                        file1.csv
                        file2.csv
                        ...
                LBNL/
                    clean.csv
                    fault/
                        file1.csv
                        file2.csv
                        ...
                RMFD/
                    clean/
                        file1.mat
                        file2.mat
                        ...
                    fault/
                        file1.mat
                        file2.mat
                        ...
                SKAB/
                    clean.csv
                    fault1/
                        file1.csv
                        file2.csv
                        ...
                    fault2/
                        file1.csv
                        file2.csv
                        ...
        Results structure:
            - TEP and LBNL:
                clean: {DataFrame
                fault: {filename: DataFrame, ...}
            
            - RMFD:
                clean: { {filename: DataFrame, ...} }
                fault: { {filename: DataFrame, ...} }
            
            - SKAB:
                clean: {DataFrame}
                fault: { {fault_folder: {filename: DataFrame, ...}}, ... }
    """
    results = {
        'clean': {},
        'fault': {}
    }

    dataset_path = os.path.join(data_path, dataset)
    
    # Also drop columns from the datasets. Columns are numerically given
    if dataset in ['TEP', 'LBNL']:
        
        if dataset == 'TEP':
            df_clean = pd.read_csv(os.path.join(dataset_path, 'clean.csv'))
            df_clean_test = pd.read_csv(os.path.join(dataset_path, 'clean_test.csv'))
            results['clean']['clean'] = df_clean
            results['clean']['clean_test'] = df_clean_test
        else:
            df_clean = pd.read_csv(os.path.join(dataset_path, 'clean.csv'))
            results['clean'] = df_clean
        
        fault_dir = os.path.join(dataset_path, 'fault')
        fault_files = [os.path.join(fault_dir, f) for f in sorted(os.listdir(fault_dir))]
        results['fault'] = read_files_threaded(fault_files)

    elif dataset == 'RMFD':
        clean_dir = os.path.join(dataset_path, 'clean')
        fault_dir = os.path.join(dataset_path, 'fault')

        clean_files = [os.path.join(clean_dir, f) for f in sorted(os.listdir(clean_dir)) if f.endswith(('.csv', '.mat'))]
        fault_files = [os.path.join(fault_dir, f) for f in sorted(os.listdir(fault_dir)) if f.endswith(('.csv', '.mat'))]

        results['clean'] = read_files_threaded(clean_files)
        results['fault'] = read_files_threaded(fault_files)

    elif dataset == 'SKAB':
        clean_path = os.path.join(dataset_path, 'clean.csv')
        df_clean = pd.read_csv(clean_path, sep=';')

        results['clean'] = df_clean

        for fault_dir in sorted(os.listdir(dataset_path)):
            fault_path = os.path.join(dataset_path, fault_dir)
            if os.path.isdir(fault_path) and fault_dir != 'clean.csv':
                fault_files = [os.path.join(fault_path, f) for f in sorted(os.listdir(fault_path)) if f.endswith('.csv')]
                results['fault'][fault_dir] =  read_files_threaded(fault_files)

    elif dataset == 'VSD':
        clean_root = os.path.join(dataset_path, 'clean')
        fault_root = os.path.join(dataset_path, 'fault')

        for case in sorted(os.listdir(clean_root)):
            case_path = os.path.join(clean_root, case)
            if os.path.isdir(case_path):
                clean_files = [os.path.join(case_path, f) for f in sorted(os.listdir(case_path)) if f.endswith('.txt')]
                results['clean'][case]= read_files_threaded(clean_files)

        for case in sorted(os.listdir(fault_root)):
            case_path = os.path.join(fault_root, case)
            if os.path.isdir(case_path):
                fault_files = [os.path.join(case_path, f) for f in sorted(os.listdir(case_path)) if f.endswith('.txt')]
                results['fault'][case] = read_files_threaded(fault_files)

    else:
        raise ValueError(f"Unsupported dataset: {dataset}")

    return results
````

````python
# Plotting - CUSUM Statistic

def plot_cusum_stat(stat, detections=None, h=None, title="CUSUM Statistic"):
    """
    Plot CUSUM statistic over time with optional detections and threshold.
    
    Parameters:
        stat: CUSUM statistic array
        detections: binary array, same length as stat
        h: threshold value
    """
    plt.figure(figsize=(12, 4))
    plt.plot(stat, label="CUSUM statistic", color="blue")
    if h is not None:
        plt.axhline(h, color="red", linestyle="--", label=f"Threshold h={h}")
    if detections is not None:
        anomaly_points = np.where(detections == 1)[0]
        plt.scatter(anomaly_points, stat[anomaly_points], color="orange", label="Detections", zorder=5)
    plt.title(title)
    plt.xlabel("Time")
    plt.ylabel("CUSUM Statistic")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.savefig(program_parameters["plot_path"] + title + "_stat.pdf", format='pdf')
    plt.show()
````

````python
# Plotting - for debugging
def plot_time_series(x, labels=None, title="Time Series", dataset = ""):
    """ 
        Plot a time series with optional labels for anomalies
    """
    plt.figure(figsize=(12, 5))
    
    plt.plot(x, label="Time Series", color="gray")
    plt.fill_between(range(len(labels)), x.min(), x.max(), where=labels == 1, 
                     color='red', alpha=0.2, label="True Anomaly")
    plt.title(dataset)
    plt.legend()
    plt.grid(True)

    plt.tight_layout()
    plt.savefig(program_parameters["plot_path"] + dataset + "_time_series.pdf", format='pdf')
    plt.show()
````

#### **Metrics Functions**

````python
# Metrics - TPR, FPR, DD  

def compute_metrics(lbl: np.ndarray, det: np.ndarray) -> Dict[str, float]:

    positives = lbl == 1
    negatives = lbl == 0

    tpr = np.sum(det[positives]) / np.sum(positives) if np.sum(positives) > 0 else 0
    fpr = np.sum(det[negatives]) / np.sum(negatives) if np.sum(negatives) > 0 else 0

    dd = -1
    if np.any(positives):
        first_anomaly = np.argmax(positives)
        after_anomaly = det[first_anomaly + 1:]
        detections_after = np.where(after_anomaly == 1)[0]
        if len(detections_after) > 0:
            dd = detections_after[0] + 1  # +1 since offset from anomaly

    return {'TPR': tpr, 'FPR': fpr, 'DD': dd}
    
def tpr_fpr_dd(
    labels: np.ndarray,
    detections: np.ndarray,
    simulation: bool = False,
    sim_len: Union[int, List[int]] = 960
) -> Dict[str, float]:
    
    """
    Computes True Positive Rate (TPR), False Positive Rate (FPR), and Detection Delay (DD).
    
    Parameters
    ----------
    labels : np.ndarray
        Binary array indicating ground truth anomalies.
    detections : np.ndarray
        Binary array indicating detected anomalies.
    simulation : bool, optional
        If True, treats labels/detections as concatenated simulations.
    sim_len : int or List[int], optional
        Simulation length(s). Default is 960.

    Returns
    -------
    dict
        Dictionary with keys 'TPR', 'FPR', 'DD'.
    """

    if simulation:
        TPRs, FPRs, DDs = [], [], []

        # Handle variable simulation lengths
        sim_lengths = sim_len if isinstance(sim_len, list) else [sim_len] * (len(labels) // sim_len)
        start = 0

        for length in sim_lengths:
            end = min(start + length, len(detections))
            lbl_window = labels[start:end]
            det_window = detections[start:end]

            metrics = compute_metrics(lbl_window, det_window)
            TPRs.append(metrics['TPR'])
            FPRs.append(metrics['FPR'])
            if metrics['DD'] != -1:
                DDs.append(metrics['DD'])

            start += length
            if start >= len(detections):
                break

        return {
            'TPR': float(np.mean(TPRs)),
            'FPR': float(np.mean(FPRs)),
            'DD': float(np.mean(DDs)) if DDs else -1
        }

    else:
        return compute_metrics(labels, detections)
````

````python
# Metrics - Robustness

def robustness_metric(stat = None, threshold = 0, shift_amplitudide=0):
    """
    Calculate the robustness metric 
        for a given CUSUM statistic and threshold.

    - Noise experiments (low amplitude noise.)
    - Micro shifts (micro shifts in the time series, 0.1 to 2 sigma)
    - Noise and outliers robustness score (NORS): the relative drop in performance compared to the baseline scenario 
            (e.g., on the clean dataset without noise and/or outliers)
    
    Arguments
    ---------
    stat: CUSUM statistic array (CUSUM computed values over a time series)
    threshold: threshold value
    shift_amplitudide: amplitude of the shift (noise!)

    Returns
    -------
    FAR_N: False Alarm Rate 
    
    """

    if stat is None:
        raise ValueError("CUSUM statistic is None")
    if threshold is None:
        raise ValueError("Threshold is None")

  
    FAR_N = np.sum(stat) / len(stat)

    return FAR_N
````

````python
# Metrics - Resilience

def compute_stability_time(stat_pos = None, 
                          stat_neg = None, 
                          threshold = 0, 
                          last_anomaly = 0,
                           N=10):
    """
    Computes stability time: 
         The time until the CUSUM stats stay within threshold for N consecutive steps.
    """

    
    if stat_pos is None:
        raise ValueError("CUSUM statistic is None")
    
    if threshold is None:
        raise ValueError("Threshold is None")
    
    if last_anomaly is None:
        raise ValueError("Last anomaly in data is None")
    
    data_length = len(stat_pos)

    if stat_neg is not None:
        for i in range(last_anomaly + 1, data_length - N):
            stable = True
            for j in range(N):
                if stat_pos[i + j] > threshold or stat_neg[i + j] < -threshold:
                    stable = False
                    break
            if stable:
                return i - last_anomaly 
    else:
        for i in range(last_anomaly + 1, data_length - N):
            stable = True
            for j in range(N):
                if stat_pos[i + j] > threshold:
                    stable = False
                    break
            if stable:
                return i - last_anomaly
            
    return -1
````

````python
# Metrics - Resilience

def compute_recovery_time(stat_pos = None, 
                          stat_neg = None, 
                          threshold = 0, 
                          last_anomaly_in_data = 0):
    """
    Compute the recovery time for a given anomaly.
    # After the last anomaly in data (last_anomaly_in_data),
    #  the time needed to recover below the threshold.
    """
    if stat_pos is None:
        raise ValueError("CUSUM statistic is None")
    
    if threshold is None:
        raise ValueError("Threshold is None")
    
    if last_anomaly_in_data is None:
        raise ValueError("Last anomaly in data is None")
    
    last_anomaly_in_cusum = 0

    if stat_neg is not None:
        for i in range(last_anomaly_in_data + 1, len(stat_pos)):
            if  (stat_pos[i] <= threshold and stat_neg[i] >= -threshold):
                last_anomaly_in_cusum = i
                break
    else:
        for i in range(last_anomaly_in_data + 1, len(stat_pos)):
            if  (stat_pos[i] <= threshold):
                last_anomaly_in_cusum = i
                break

    recovery_time = last_anomaly_in_cusum - last_anomaly_in_data
    
    rec_time = -1 if last_anomaly_in_cusum == 0 else recovery_time

    return rec_time
````

````python
# Metrics - Resilience

def resilience_metric(stat_pos = None, 
                      stat_neg = None,
                      threshold = 0, 
                      last_anomaly_in_data = 100, 
                      simulation = False,
                      sim_length = 1000,
                      N = 10):

    """
    Calculate the resilience metric for a given CUSUM statistic and threshold.

    After the last anomaly, the time needed to recover and return below the threshold.

    Recovery Time: 
    The time it takes for the CUSUM statistic to recover below the threshold 
                after the last anomaly.

    Stability time:
        The time until the CUSUM stats stays within threshold for N consecutive steps.
    
    Arguments
    ---------
    stat: CUSUM statistic array (CUSUM computed values over a time series)
    threshold: threshold value
    last_anomaly_in_data: index of the last anomaly in the data
    alpha: parameter for the resilience score

    returns
    -------
    recovery_time: the time it takes for the CUSUM statistic to recover below the threshold 
                after the last anomaly.
    stability_time: the time it takes for the CUSUM statistic to recover and stay below the threshold 
               for N samples after the last anomaly.
    """

    if stat_pos is None or len(stat_pos) == 0:
        raise ValueError("CUSUM statistic is None or empty")

    if last_anomaly_in_data is None:
        raise ValueError("Last anomaly in data is None")
    
    if threshold is None:
        raise ValueError("Threshold is None")
    
    if last_anomaly_in_data is None:
        raise ValueError("Last anomaly in data is None")
    

    # if we are working with simulations we need to compute for each sim
    
    if simulation:       
        
        recovery_times = []
        stability_times = []
        num_sims = len(stat_pos) // sim_length 

        # Trim the data to the number of sims
        stat_pos = stat_pos[:num_sims * sim_length]
        stat_pos_2D = stat_pos.reshape((num_sims, sim_length))
        
        if stat_neg is not None:
            stat_neg = stat_neg[:num_sims * sim_length] 
            stat_neg_2D = stat_neg.reshape((num_sims, sim_length))

            # Compute for each sim
            for sim_pos_row, sim_neg_row in zip(stat_pos_2D, stat_neg_2D):
                r_time = compute_recovery_time(sim_pos_row, 
                                            sim_neg_row, 
                                            threshold, 
                                            last_anomaly_in_data)
                
                s_time = compute_stability_time(sim_pos_row, 
                                                sim_neg_row, 
                                                threshold, 
                                                last_anomaly_in_data, 
                                                N=N)

                recovery_times.append(r_time)
                stability_times.append(s_time)
        else:
            # Compute for each sim
            for sim_pos_row in stat_pos_2D:
                r_time = compute_recovery_time(sim_pos_row, 
                                            None, 
                                            threshold, 
                                            last_anomaly_in_data)
                
                s_time = compute_stability_time(sim_pos_row, 
                                                None, 
                                                threshold, 
                                                last_anomaly_in_data, 
                                                N=10)

                recovery_times.append(r_time)
                stability_times.append(s_time)

        # Compute the mean without -1 or 0 values in the array
        recovery_times = np.array(recovery_times)
        recovery_times = recovery_times[recovery_times != -1]
        recovery_times = recovery_times[recovery_times != 0]

        stability_times = np.array(stability_times)
        stability_times = stability_times[stability_times != -1]
        stability_times = stability_times[stability_times != 0]

        recovery_time = np.mean(recovery_times)
        stability_time = np.mean(stability_times)

    return recovery_time, stability_time
````

#### **Cumulative Sums Functions - CUSUM**

````python
# CUSUM - Hampel Function
def hampel(e, a, b, r, gamma):
    """ 
    Hampel function as it appears in [4] 
    See also: Huber function.
    
    arguments
    ---------
    e: error
    a, b, r, gamma: hampel function parameters

    returns
    -------
    Hampel function value
    """
    
    c = (a/2) * (b - a + r)

    if abs(e) <= a:
        return (e**2)/(2*c)
    
    elif abs(e) > a and abs(e) <= b:
        return ((1/2)*(a**2) + a*(abs(e) - a))/c
    
    elif abs(e) > b and abs(e) <= gamma:
        return (a/2)*(2*b - a + (abs(e) - b)*(1 + (r - abs(e)/(c-b)))) / c
    
    elif (abs(e) > gamma):
        return 1
````

````python
# CUSUM - Classical Page's CUSUM (The father of CUSUMs)
def classical_cusum(x, 
                    target_mean = 0, 
                    k=0.5, 
                    h=5,
                    simulation = False,
                    sim_length = 960,
                    resilience = False):
    """
        Input:
            x: input time series
            target_mean: expected mean under normal conditions
            std_dev: standard deviation under normal conditions
            k: reference value (usually 0.5 * shift)
            h: threshold
        Returns: 
            pos: positive CUSUM statistic
            neg: negative CUSUM statistic
            detections: binary vector
    """
    # Input validation
    x = np.asarray(x, dtype=float)
    if x.size == 0:
        raise ValueError("Input array cannot be empty")
    
    n = len(x)
    
    # Initialize arrays
    pos_cusum = np.zeros(n)
    neg_cusum = np.zeros(n)
    detections = np.zeros(n, dtype=bool)
    
    # Handle simulation reset points
    reset_points = set()
    if simulation and not resilience:
        if isinstance(sim_length, list):
            reset_points = set(np.cumsum(sim_length))
        else:
            reset_points = set(range(sim_length, n, sim_length))
    
    # Main CUSUM computation
    for i in range(1, n):
        if i in reset_points:
            pos_cusum[i] = 0
            neg_cusum[i] = 0
            continue
            
        deviation = (x[i] - target_mean) 
        
        pos_cusum[i] = max(0, pos_cusum[i-1] + deviation - k)
        neg_cusum[i] = min(0, neg_cusum[i-1] - deviation + k)
    
        # detections[i] = int(pos_cusum[i] > h or neg_cusum[i] < -h)

    detections = (pos_cusum > h) | (neg_cusum < -h)
    detections = detections.astype(np.int32)
    
    return pos_cusum, neg_cusum, detections
````

````python
# CUSUM - 1-CUSUM Scheme
def one_cusum_scheme(x, 
                     target_mean, 
                     lamb = 0.5,
                     k = 0.5, 
                     h = 5,
                     simulation = False,
                     sim_length = 960,
                     resilience = False):
    """
        Input:
            x: input time series
            target_mean: expected mean under normal conditions
            lamb: smoothing factor
            k: reference value (usually 0.5 * shift)
            h: threshold
        Returns: 
            s: CUSUM statistic
            detections: binary vector
    """


    x = np.asarray(x)
    n = len(x)

    s = np.zeros(n)
    detections = np.zeros(n, dtype=np.bool)

    y = x - target_mean
    y_square = y ** 2

    # Handle simulation segmentation
    reset_points = set()
    if simulation and not resilience:
        if isinstance(sim_length, list):
            reset_points = set(np.cumsum(sim_length))
        else:
            reset_points = set(range(sim_length, n, sim_length))

    for i in range(1, n):
        if i in reset_points:
            s[i] = 0
            continue

        # Paper-defined CUSUM update logic
        if s[i-1] > 0 or (s[i-1] == 0 and y[i] > 0):
            s[i] = max(0, s[i-1] + lamb * y[i] + (1 - lamb) * y_square[i] - k)
        else:
            s[i] = min(0, s[i-1] + lamb * y[i] - (1 - lamb) * y_square[i] + k)

        # Detection logic
        #detections[i] = 1 if abs(s[i]) > h else 0

    detections = (np.abs(s) > h)
    detections = detections.astype(np.int32)

    return s, detections
````

````python
# CUSUM - ACUSUM Hampel control chart
def hampel_acusum(x, 
                  a, 
                  b, 
                  r = 3, 
                  gamma = 4.5,
                  h=5, 
                  k_plus = 0.5, 
                  k_minus = 0.5,
                  simulation = False,
                  sim_length = 960,
                  resilience = False,):
    """ As it appears in [4].
        Input:
            x: input time series
            a, b, r, gamma: hampel parameters
            k +/- : reference values
            h: threshold
        Returns: 
            pos, neg: CUSUM positive and negative statistics
            detections: binary vector

    *** 
    If someone accidentally reads this:
    Please for the love of all that is holy, when you describe an algorithm
    in your paper, describe it clearly enough so that a person can implement it.
    Authors in [4], I'm looking at you ...

    P.S. r parameter is forgotten in the paper, freestyling it...
    Equation 9 also mentions a "d" parameter (ghost parameter).
    The referenced source of the Hampel function is non existent.
    """    

    x = np.asarray(x)
    n = len(x)

    pos = np.zeros(n)
    neg = np.zeros(n)
    detections = np.zeros(n, dtype=np.bool)

    # Precompute simulation skip points
    if simulation and not resilience:
        if isinstance(sim_length, list):
            sim_points = set(np.cumsum(sim_length))
        else:
            sim_points = set(range(sim_length, n, sim_length))
    else:
        sim_points = set()

    for i in range(1, n):
        if i in sim_points:
            pos[i] = 0
            neg[i] = 0
            continue

        # Positive direction
        eip = x[i] - pos[i-1]
        wip = hampel(eip, a, b, r, gamma) / eip if eip != 0 else 0
        delta_plus = max(wip, k_plus)
        pos[i] = max(0, pos[i-1] + delta_plus * (x[i] - delta_plus / 2))

        # Negative direction
        eim = x[i] - neg[i-1]
        wim = hampel(eim, a, b, r, gamma) / eim if eim != 0 else 0
        delta_minus = min(wim, k_minus)
        neg[i] = min(0, neg[i-1] - delta_minus * (x[i] - delta_minus / 2))

        # Detection check
        # detections[i] = 1 if pos[i] > h or neg[i] < -h else 0
        
    detections = (pos > h) | (neg < -h)
    detections = detections.astype(np.int32)

    return pos, neg, detections
````

````python
# CUSUM - High Order CUSUM based on Skewness and Kurtosis
def high_order_cusum(x, 
                     window = 30, 
                     alpha = 0.5, 
                     exp_x = 0, 
                     k = 0, 
                     h = 5,
                     simulation = False,
                     sim_length = 960,
                     resilience = False):
    """
        High Order CUSUM based on Skewness and Kurtosis
        Input:
            x: input time series
            k: reference value (usually 0.5 * shift)
            h: threshold
        Returns: 
            s: CUSUM statistic
            detections: binary vector
    """
    x = np.asarray(x)
    n = len(x)

    pos = np.zeros(n)
    neg = np.zeros(n)
    s = np.zeros(n)
    detections = np.zeros(n, dtype=np.bool)

    x = pd.Series(x)

    # Precompute simulation breaks
    if simulation and not resilience:
        if isinstance(sim_length, list):
            sim_points = set(np.cumsum(sim_length))
        else:
            sim_points = set(range(sim_length, n, sim_length))
    else:
        sim_points = set()

    for i in range(1, n):
        if i < window or i in sim_points:
            pos[i] = 0
            neg[i] = 0
            continue

        window_data = x[i - window:i]
        std_val = np.std(window_data)

        if std_val > 1e-10:
            skewness_w = window_data.skew(skipna = True)
            kurtosis_w = window_data.kurtosis(skipna = True)
        else:
            skewness_w = 0
            kurtosis_w = 0

        z = alpha * skewness_w + (1 - alpha) * kurtosis_w

        val_pos = max(0, pos[i-1] + z - exp_x - k)
        val_neg = min(0, neg[i-1] - z + exp_x + k)

        pos[i] = val_pos
        neg[i] = val_neg
        s[i] = max(abs(val_pos), abs(val_neg))

        #detections[i] = 1 if val_pos > h or val_neg < -h else 0
    detections = (pos > h) | (neg < -h)
    detections = detections.astype(np.int32)

    return pos, neg, s, detections
````

````python
# CUSUM - High Order CUSUM based on Skewness and Kurtosis
def high_order_cusum_params(x, 
                     window = 30, 
                     alpha = 0.5, 
                     exp_x = 0, 
                     k = 0, 
                     h = 5,
                     simulation = False,
                     sim_length = 960,):
    """
        High Order CUSUM based on Skewness and Kurtosis
        Input:
            x: input time series
            k: reference value (usually 0.5 * shift)
            h: threshold
        Returns: 
            s: CUSUM statistic
            detections: binary vector
    """
    n = len(x)

    pos = np.zeros(n)
    neg = np.zeros(n)
    s = np.zeros(n)
    detections = np.zeros(n, dtype=np.bool_)

    if simulation:
        if isinstance(sim_length, list):
            sim_points = set(np.cumsum(sim_length))
        else:
            sim_points = set(range(sim_length, n, sim_length))
    else:
        sim_points = set()
    
    x = pd.Series(x)
    skews = x.rolling(window=window).skew()
    kurts = x.rolling(window=window).kurt()

    z = alpha * skews + (1 - alpha) * kurts
    z = z.fillna(0).to_numpy()   
    
    # cusum recursion
    for i in range(1, n):
        if i in sim_points:
            pos[i] = 0
            neg[i] = 0
            continue

        pos[i] = max(0, pos[i-1] + z[i] - exp_x - k)
        neg[i] = min(0, neg[i-1] - z[i] + exp_x + k)

        s[i] = max(abs(pos[i]), abs(neg[i]))

    detections = (pos > h) | (neg < -h)
    detections = detections.astype(np.int32)

    return pos, neg, s, detections
````

````python
# CUSUM - UCUSUM mean shift
def UCUSUM(x, 
           window = 30, 
           target_mean = 0,
           h=5,
           simulation = False,
           sim_length = 960,
           resilience = False):
    """ UCUSUM as defined in [6] 
        Input:
            x: input time series
            window: window size
            h: threshold
            target_mean: expected mean under normal conditions
        Returns: 
            pos, neg: CUSUM positive and negative statistics
            detections: binary vector
    """

    x = np.asarray(x)
    n = len(x)

    s = np.zeros(n)
    detections = np.zeros(n)

    # Precompute simulation break points for skipping
    if simulation and not resilience:
        if isinstance(sim_length, list):
            sim_points = set(np.cumsum(sim_length))
        else:
            sim_points = set(range(sim_length, n, sim_length))
    else:
        sim_points = set()

    for i in range(window, n):
        if i in sim_points:
            s[i] = 0
            continue

        local_mean = np.mean(x[i-window:i])
        val = local_mean - target_mean

        s[i] = val
        # detections[i] = 1 if val > h else 0

    detections = (s > h)
    detections = detections.astype(np.int32)
    
    return s, detections
````

````python
# CUSUM - Variance-change CUSUM 
def variance_cusum(x, 
                   target_std, 
                   window = 20,
                   k = 0.5, 
                   h = 5,
                   simulation = False,
                   sim_length = 960,
                   resilience = False):
    """
        Input:
            x: input time series
            window: window size
            h: threshold
            k: reference value
            target_std: expected std under normal conditions
        Returns: 
            pos, neg, s: CUSUM positive, negative and total statistics
            detections: binary vector
    """
    x = np.asarray(x)
    n = len(x)
    
    pos = np.zeros(n)
    neg = np.zeros(n)
    detections = np.zeros(n)

    if simulation and not resilience:
        if isinstance(sim_length, list):
            sim_points = set(np.cumsum(sim_length))
        else:
            sim_points = set(range(sim_length, n, sim_length))
    else:
        sim_points = set()

    for i in range(window, n):
        if i in sim_points:
            pos[i] = 0
            neg[i] = 0
            continue
        
        local_std = np.std(x[i-window:i])

        z = local_std - target_std

        pos[i] = max(0, pos[i-1] + z - k)
        neg[i] = min(0, neg[i-1] + z + k)  
        
        #detections[i] = 1 if pos[i] > h or neg[i] < -h else 0

    detections = (pos > h) | (neg < -h)
    detections = detections.astype(np.int32)

    return pos, neg, detections
````

````python
# CUSUM Utilities - Threshold Estimation

def estimate_cusum_threshold(cusum_stats_pos = None, 
                             cusum_stats_neg = None,
                             clean_data = None, 
                             h = 5,
                             false_alarm_rate = 0.01,
                             csm_type = "classical",
                             window = 30,
                             alpha  = 0.5,
                             max_threshold = False):
    """
    Estimate threshold h from normal data CUSUM stats.
    
    Input:
        cusum_stats: array of CUSUM statistics under normal data
        clean_data: array of clean data (optional)
        h: threshold constant
        
    Returns:
        finial threshold value
    """
    if cusum_stats_pos is not None and cusum_stats_neg is not None:
        cusum_stats = abs(np.concatenate((cusum_stats_pos, cusum_stats_neg)))
    else:
        cusum_stats = cusum_stats_pos
        
    if clean_data is None:
        if max_threshold:
            return max(cusum_stats)
        return np.mean(cusum_stats) + h * np.std(cusum_stats)
    else:
        if max_threshold:
            return max(clean_data)
        return h * np.std(clean_data)
````

````python
# CUSUM Utilities - Estimate target mean from cusum over normal data
def estimate_target_mean(x):
    return np.mean(x)
````

````python
# CUSUM Utilities - Reference Value Estimation

def estimate_k_from_shift(std_dev, shift_size=1.0):
    """
    Choose the k (reference value) for CUSUM as a fraction of std deviation.
    
    std_dev: baseline standard deviation if either clean data or CUSUM on clean data
    shift_size: expected shift in mean, in units of std_dev (e.g., 1 sigma)
        
    """
    return shift_size * std_dev
````

````python
def estimate_k_from_stats(data, alpha=0.5, shift_size=1.0, window=30):
    """
    Estimate k (reference value) from statistics.
    Uses rolling window skewness and kurtosis instead of manual loops.
    """
    data = pd.Series(data)

    skews = data.rolling(window=window).skew()
    kurts = data.rolling(window=window).kurt()

    valid_mask = ~skews.isna() & ~kurts.isna()
    skews = skews[valid_mask].to_numpy()
    kurts = kurts[valid_mask].to_numpy()

    combined_stats = alpha * skews + (1 - alpha) * kurts

    k_estimated = np.std(combined_stats) * shift_size
    exp_x = np.mean(combined_stats)

    return exp_x, k_estimated
````

````python
# Reshape data into sliding windows

def create_sliding_windows(data, window_size = 30):
    
    """
    Parameters:
    - data: np.ndarray, shape (n_samples,) or (n_samples, n_features)
    - window_size: int, size of each window

    Returns:
    - windows: np.ndarray, shape (n_windows, window_size * n_features)
    """

    if data.ndim == 1:
        data = data.reshape(-1, 1)
    
    n_samples, n_features = data.shape
    windows = []

    for i in range(n_samples - window_size + 1):
        window = data[i:i+window_size].flatten()
        windows.append(window)

    return np.array(windows)
````

````python
# Create ML Models
def create_model(model_parameters = None, 
                 number_inputs = 1, 
                 number_outputs = 1):
    """
        Create a keras recurrent neural object from a dictionary of parameters.
    """

    if model_parameters is None:
        raise ValueError("Model parameters must be provided.")
    if number_inputs is None or number_inputs == 0:
        raise ValueError("Number of inputs must be provided.")
    if number_outputs is None or number_outputs == 0:
        raise ValueError("Number of outputs must be provided.")
    
    # TF for LSTMTF
    if model_parameters['TF'] == True:
        number_inputs = number_inputs + 1
        
    # Set random seed
    keras.utils.set_random_seed(model_parameters['seed'])

    # In newer versions of Keras, for stateful LSTM, you need to specify the batch_input_shape as the first layer (input layer)
    model = keras.Sequential()

    
    model.add(keras.layers.InputLayer(batch_input_shape=(model_parameters['batch_size'], 
                                                        model_parameters['seq_len'], 
                                                        number_inputs)))
    
    for i in range(model_parameters['num_layers']): 
        if model_parameters['models'] == 'RNN':
            model.add(keras.layers.SimpleRNN(units=model_parameters['hidden_size'][i], 
                                             return_sequences=True,
                                             stateful=model_parameters['stateful']))

        if model_parameters['models'] == 'LSTM':
            model.add(keras.layers.LSTM(units = model_parameters['hidden_size'][i],
                                            return_sequences=True, 
                                            stateful=model_parameters['stateful']))

        if model_parameters['models'] == 'GRU':
            model.add(keras.layers.GRU(units = model_parameters['hidden_size'][i],
                                            return_sequences=True,
                                            stateful=model_parameters['stateful']))

    model.add(keras.layers.Dense(number_outputs))

    model.compile(optimizer=model_parameters['optimizer'], 
                  loss=model_parameters['loss'], 
                  metrics=model_parameters['metrics'])

    # https://keras.io/api/optimizers/

    return model
````

````python
# Measure time for model training and prediction
def measure_model_time( clean_data = None,
                        fault_data = None,
                        datasets_parameters = None,
                        model = None,
                        models_parameters = None,
                        LENGTH = 0,
                        dataset = 'TEP',
                        data = None,):
    
    """
        Load datasets, normalize data, train model and get residuals
        Arguments
        ----------
        clean_data: clean data to use for training
        fault_data: fault data to use for testing
        datasets_parameters: dictionary of datasets parameters
        model: keras model to use for training
        models_parameters: dictionary of model parameters
        LENGTH: length of the data to use for training

        Returns
        -------
        time_measure: time taken for model training and prediction
    """
    datasets_parameters = datasets_parameters[dataset]
    dropped_columns = datasets_parameters['dropped_columns']
    files = datasets_parameters['files']
    
    clean_raw, fault_raw, labels_clean_raw, labels_fault_raw, sim_length = \
                    get_clean_fault_data(dataset = dataset, 
                                        data = data, 
                                        param_estimation_size = PARAM_ESTIMATION_SIZE, 
                                        max_length = MAX_LENGTH,
                                        dropped_columns = dropped_columns,
                                        start = 0,
                                        files = files)
    

    clean_data = clean_raw.copy()
    fault_data = fault_raw.copy()
    
    results_time_measure = { }
    results_time_measure_fault_data = []

    # Normalize
    scaler = MinMaxScaler()
    scaler.fit(clean_data)
    clean_data = scaler.transform(clean_data)
    clean_data = clean_data[0:LENGTH, :]

    # save_pickle(clean_data, 'clean_data' + str(LENGTH) + 'rpi.pkl')

    # Clean Data Experiments
    X_clean, y_clean = clean_data[:,datasets_parameters['inputs']], clean_data[:,datasets_parameters['outputs']]
    
    if models_parameters['TF']:
        X_clean = np.hstack((X_clean[1:, :], y_clean[0:-1]))
        y_clean = y_clean[1:, :]

    seq_len = models_parameters['seq_len']
    
    if models_parameters['TF']:
        num_feats = len(datasets_parameters['inputs']) + 1
    else:
        num_feats = len(datasets_parameters['inputs'])
        
    num_outputs = len(datasets_parameters['outputs'])

    valid_rows_clean = X_clean.shape[0] // seq_len * seq_len

    if len(y_clean.shape) == 1:
        X_clean = X_clean[:valid_rows_clean]
        y_clean = y_clean[:valid_rows_clean]
    else:
        X_clean = X_clean[:valid_rows_clean, :]
        y_clean = y_clean[:valid_rows_clean, :]

    num_sequences_clean = X_clean.shape[0] // seq_len

    X_clean = X_clean.reshape(num_sequences_clean, seq_len, num_feats)
    y_clean = y_clean.reshape(num_sequences_clean, seq_len, num_outputs)

    # If the model is statefull, trim to the maximum multiples of batch size
    if models_parameters['stateful']:
        maximum_sequences = X_clean.shape[0] - X_clean.shape[0] % models_parameters['batch_size']
        X_clean = X_clean[:maximum_sequences]
        y_clean = y_clean[:maximum_sequences]

    # Measure time for model training
    start_time_train = time.perf_counter()

    model.fit(X_clean, 
              y_clean, 
              epochs = models_parameters['epochs'], 
              batch_size = models_parameters['batch_size'],
              verbose = 0, 
              shuffle = False)
    
    end_time_train = time.perf_counter()
    results_time_measure['train'] = end_time_train - start_time_train


    fault_data = scaler.transform(fault_data)
    fault_data = fault_data[0:LENGTH, :]
    #save_pickle(fault_data, 'faulty_data' + str(LENGTH) + 'rpi.pkl')

    X_fault, y_fault = fault_data[:,datasets_parameters['inputs']], fault_data[:,datasets_parameters['outputs']]

    if models_parameters['TF']:
        X_fault = np.hstack((X_fault[1:, :], y_fault[0:-1]))
        y_fault = y_fault[1:, :]

    valid_rows_fault = X_fault.shape[0] // seq_len * seq_len

    if len(y_fault.shape) == 1:
        X_fault = X_fault[:valid_rows_fault]
        y_fault = y_fault[:valid_rows_fault]
    else:
        X_fault = X_fault[:valid_rows_fault, :]
        y_fault = y_fault[:valid_rows_fault, :]
    
    num_sequences_fault = X_fault.shape[0] // seq_len
    
    X_fault = X_fault.reshape(num_sequences_fault, seq_len, num_feats)
    y_fault = y_fault.reshape(num_sequences_fault, seq_len, num_outputs)

    if models_parameters['stateful']:
        maximum_sequences = X_fault.shape[0] - X_fault.shape[0] % models_parameters['batch_size']
        X_fault = X_fault[:maximum_sequences]
        y_fault = y_fault[:maximum_sequences]
    
    start_time_predict = time.perf_counter()

    model.predict(X_fault, batch_size = models_parameters['batch_size'], verbose = 0)
    
    end_time_predict = time.perf_counter()
    results_time_measure_fault_data.append(end_time_predict - start_time_predict)

    results_time_measure['test'] = results_time_measure_fault_data

    return results_time_measure
````

````python
# Get Data From Loaded Pikles
def get_clean_fault_data(dataset = None, 
                         data = None, 
                         param_estimation_size = 0.1, 
                         max_length = 0,
                         files = None,
                         dropped_columns = None,
                         start = 0
                         ):
    """
        Get Clean and Fault Data from the data dictionary

        Arguments
        ----------
        dataset: dataset to get data from
        data: dictionary of data
        param_estimation_size: size of data to use for CUSUM parameter estimation
        max_length: maximum length of data to use
        files: list of files to use
        dropped_columns: list of columns to drop (e.g., time, label, etc.)

        Returns
        -------
        List of np arrays: clean and fault labels, clean and fault data
    """
    # if data is None:
    #     data = load_pickle('data.pkl')

    number_of_faults = 0

    if dataset is None:
        dataset = program_parameters['dataset']
        
    datafile = "datasets/np_all_data_" + str(dataset) + "_" + str(max_length) + ".pkl"


    if os.path.exists(datafile):
        temp_data = load_pickle(datafile)
        
        clean = temp_data['clean']
        fault = temp_data['fault']
        labels_clean = temp_data['labels_clean']
        labels_fault = temp_data['labels_fault']
        sim_length = temp_data['sim_length']

        return clean, fault, labels_clean, labels_fault, sim_length

    else:
        temp_data = {"clean": None, 
                "fault": None,
                "labels_clean": None,
                "labels_fault": None,
                "sim_length": None}
        

        # Create DATA for each dataset
        if dataset == "TEP":
            
            faults_size = 0

            # If all flag is selected, return all data from all dictionaries
            # It will be returned as a list of dataframes

            if files == 'all':
                fault = []
                clean = data['TEP']['clean']['clean'].values
                for key in data['TEP']['fault'].keys():
                    vals = data['TEP']['fault'][key].values

                    if max_length > 0:
                        vals = vals[:max_length, :]

                    fault.append(vals)

                # Create a single file with all faults
                fault = np.vstack(fault)
                faults_size = fault.shape[0]
                
                # Labels
                labels_clean = np.zeros(clean.shape[0], dtype=int)

                sim_length = 960 # All faults have the same length 960
                labels_fault = np.zeros(faults_size, dtype=int)
                block = 160 + 800  # total cycle length

                # Create pattern for TEP dataset
                for i in range(0, faults_size + 1, block):
                    anomaly_start = i + 160
                    anomaly_end = min(anomaly_start + 800, faults_size)
                    labels_fault[anomaly_start:anomaly_end] = 1

            else:
                raise Exception("Set files to 'all' for TEP dataset")


        elif dataset == "LBNL":

            if files == 'all':
                fault = []
                clean = data['LBNL']['clean'].values
                for key in data['LBNL']['fault'].keys():
                    vals = data['LBNL']['fault'][key].values

                    if max_length > 0:
                        vals = vals[:max_length, :]

                    fault.append(vals)

                # Create a single file with all faults
                sim_length = fault[0].shape[0] # All faults have the same length
                fault = np.vstack(fault)

                # Labels
                labels_clean = np.zeros(clean.shape[0], dtype=int)
                labels_fault = np.ones(fault.shape[0], dtype=int)

            else:
                raise Exception("Set files to 'all' for LBNL dataset")

        elif dataset == "SKAB":
            sim_length = []

            if files == "all":
                fault = []
                clean = data['SKAB']['clean'].values
                
                for key in data['SKAB']['fault'].keys():
                    for key2 in data['SKAB']['fault'][key].keys():
                        vals = data['SKAB']['fault'][key][key2].values

                        if max_length > 0:
                            vals = vals[:max_length, :]

                        fault.append(vals)

                labels_fault = []

                for i in range(len(fault)):
                    length = fault[i].shape[0]
                    labels = fault[i][:, -2]
                    labels_fault.append(np.int32(labels[:length]))
                    fault[i] = fault[i][:length, :-2]
                    sim_length.append(fault[i].shape[0])   

                # Create a single file with all faults
                fault = np.vstack(fault)

                # Labels
                labels_clean = np.zeros(clean.shape[0], dtype=int)
                labels_fault = np.hstack(labels_fault)

            else:
                raise Exception("Set files to 'all' for SKAB dataset")

        elif dataset == "RMFD":

            if files == "all":
                fault = []
                clean = []
                for key in data['RMFD']['clean'].keys():
                    clean.append(data['RMFD']['clean'][key].values)
                
                clean = np.vstack(clean)
                
                for key in data['RMFD']['fault'].keys():
                    vals = data['RMFD']['fault'][key].values

                    if max_length > 0:
                        vals = vals[:max_length, :]
                    
                    fault.append(vals)

                sim_length = fault[0].shape[0] # All faults have the same length
                fault = np.vstack(fault)

                # Labels
                labels_clean = np.zeros(clean.shape[0])
                labels_fault = np.ones(fault.shape[0], dtype=int)

        else:
                raise Exception("Set files to 'all' for RMFD dataset")
            
        # Drop Selected Columns
        if dropped_columns is not None:
            if files != 'all':
                clean = np.delete(clean, dropped_columns, axis=1)
                fault = np.delete(fault, dropped_columns, axis=1)
            else:
                clean = np.delete(clean, dropped_columns, axis=1)

                if type(fault) != list:
                    fault = np.delete(fault, dropped_columns, axis=1)
                else:
                    for i in range(len(fault)):
                        fault[i] = np.delete(fault[i], dropped_columns, axis=1)

        #Truncate Data
        if max_length != 0:
            clean = np.array(clean[0: max_length, :], dtype=float)
        else:
            clean = np.array(clean, dtype=float)

        len_clean = clean.shape[0]
        len_fault = fault.shape[0]

        labels_clean =  np.zeros(len_clean, dtype=int)
        labels_fault = labels_fault[:len_fault]
        
        # Save Data
        temp_data['clean'] = clean
        temp_data['fault'] = fault
        temp_data['labels_clean'] = labels_clean
        temp_data['labels_fault'] = labels_fault
        temp_data['sim_length'] = sim_length
        
        save_pickle(temp_data, datafile)
        
    return clean, fault, labels_clean, labels_fault, sim_length
````

````python
# Get Clean and Fault Data Residuals
def get_clean_fault_residuals_data( clean_data = None,
                                    fault_data = None,
                                    max_length = 0,
                                    datasets_parameters = None,
                                    model = None,
                                    models_parameters = None,
                                    measure_time = False,
                                    LENGTH = 0,):
    
    """
        Load datasets, normalize data, train model and get residuals
        Arguments
        ----------
        dataset: dataset to get data from
        data: dictionary of data
        datasets_parameters: dictionary of datasets parameters
        plots: whether to plot the results
        model: keras model to use for training
        models_parameters: dictionary of model parameters
        start: start index for data slicing
        
        Returns
        -------
        residuals_clean: residuals for clean data
        residuals_fault: list of residuals for fault data
        labels_clean: labels for clean data
        labels_fault: labels for fault data
        time_measure: time taken for model training and prediction
    """
    
    if measure_time:
        clean_data = clean_data[:LENGTH + LENGTH * datasets_parameters['train_ratio'], :]

    results_time_measure = { }
    results_time_measure_fault_data = []

    # Normalize
    scaler = MinMaxScaler()
    scaler.fit(clean_data)
    clean_data = scaler.transform(clean_data)

    # Clean Data Experiments
    X_clean, y_clean = clean_data[:,datasets_parameters['inputs']], clean_data[:,datasets_parameters['outputs']]
    
    if models_parameters['TF']:
        # Add TF as the previous output, as a column to the input
        X_clean = np.hstack((X_clean[1:, :], y_clean[0:-1]))
        y_clean = y_clean[1:, :]

    seq_len = models_parameters['seq_len']
    
    if models_parameters['TF']:
        num_feats = len(datasets_parameters['inputs']) + 1
    else:
        num_feats = len(datasets_parameters['inputs'])
        
    num_outputs = len(datasets_parameters['outputs'])

    valid_rows_clean = X_clean.shape[0] // seq_len * seq_len

    if len(y_clean.shape) == 1:
        X_clean = X_clean[:valid_rows_clean]
        y_clean = y_clean[:valid_rows_clean]
    else:
        X_clean = X_clean[:valid_rows_clean, :]
        y_clean = y_clean[:valid_rows_clean, :]

    num_sequences_clean = X_clean.shape[0] // seq_len

    X_clean = X_clean.reshape(num_sequences_clean, seq_len, num_feats)
    y_clean = y_clean.reshape(num_sequences_clean, seq_len, num_outputs)

    train_size = int(X_clean.shape[0] * models_parameters['train_ratio'])

    X_train_clean = X_clean[0: train_size]
    y_train_clean = y_clean[0: train_size]

    X_test_clean = X_clean[train_size: X_clean.shape[0]]
    y_test_clean = y_clean[train_size: y_clean.shape[0]]

    # If the model is statefull, trim to the maximum multiples of batch size
    if models_parameters['stateful']:
        maximum_sequences = X_train_clean.shape[0] - X_train_clean.shape[0] % models_parameters['batch_size']
        maximum_sequences_test = X_test_clean.shape[0] - X_test_clean.shape[0] % models_parameters['batch_size']
        X_train_clean = X_train_clean[:maximum_sequences]
        X_test_clean = X_test_clean[:maximum_sequences_test]
        y_train_clean = y_train_clean[:maximum_sequences]
        y_test_clean = y_test_clean[:maximum_sequences_test]

    # Measure time for model training
    if measure_time:
        start_time_train = time.perf_counter()

    model.fit(X_train_clean, 
              y_train_clean, 
              epochs = models_parameters['epochs'], 
              batch_size = models_parameters['batch_size'],
              verbose = 0, 
              shuffle = False)
    
    if measure_time:
        end_time_train = time.perf_counter()
        results_time_measure['train'] = end_time_train - start_time_train

    y_hat_clean = model.predict(X_test_clean, batch_size = models_parameters['batch_size'], verbose = 0)


    y_t_clean = y_test_clean.flatten()
    y_h_clean = y_hat_clean.flatten()

    # Get residuals
    residuals_clean = ((y_t_clean - y_h_clean)**2)

    if datasets_parameters['files'] != 'all':
        fault_data = [fault_data]
    
    residuals_fault = []
    
    if type(fault_data) == list:

        for i in range(len(fault_data)):
            # Fault Data Experiments
            fault_data[i] = scaler.transform(fault_data[i])
            
            fd = fault_data[i]

            X_fault, y_fault = fd[:,datasets_parameters['inputs']], fd[:,datasets_parameters['outputs']]

            if models_parameters['TF']:
                # Add TF as the previous output, as a column to the input
                X_fault = np.hstack((X_fault[1:, :], y_fault[0:-1]))
                y_fault = y_fault[1:, :]

            valid_rows_fault = X_fault.shape[0] // seq_len * seq_len

            if len(y_fault.shape) == 1:
                X_fault = X_fault[:valid_rows_fault]
                y_fault = y_fault[:valid_rows_fault]
            else:
                X_fault = X_fault[:valid_rows_fault, :]
                y_fault = y_fault[:valid_rows_fault, :]
            
            num_sequences_fault = X_fault.shape[0] // seq_len
            
            X_fault = X_fault.reshape(num_sequences_fault, seq_len, num_feats)
            y_fault = y_fault.reshape(num_sequences_fault, seq_len, num_outputs)

            if models_parameters['stateful']:
                maximum_sequences = X_fault.shape[0] - X_fault.shape[0] % models_parameters['batch_size']
                X_fault = X_fault[:maximum_sequences]
                y_fault = y_fault[:maximum_sequences]
            

            y_hat_fault = model.predict(X_fault, batch_size = models_parameters['batch_size'], verbose = 0)


            y_f = y_fault.flatten()
            y_hf = y_hat_fault.flatten()

            # Get residuals
            residuals_fault.append((y_f - y_hf)**2)
    else:
        fault_data = scaler.transform(fault_data)
            
        fd = fault_data
        X_fault, y_fault = fd[:,datasets_parameters['inputs']], fd[:,datasets_parameters['outputs']]

        if models_parameters['TF']:
            # Add TF as the previous output, as a column to the input
            X_fault = np.hstack((X_fault[1:, :], y_fault[0:-1]))
            y_fault = y_fault[1:, :]

        valid_rows_fault = X_fault.shape[0] // seq_len * seq_len

        if len(y_fault.shape) == 1:
            X_fault = X_fault[:valid_rows_fault]
            y_fault = y_fault[:valid_rows_fault]
        else:
            X_fault = X_fault[:valid_rows_fault, :]
            y_fault = y_fault[:valid_rows_fault, :]
        
        num_sequences_fault = X_fault.shape[0] // seq_len
        
        X_fault = X_fault.reshape(num_sequences_fault, seq_len, num_feats)
        y_fault = y_fault.reshape(num_sequences_fault, seq_len, num_outputs)

        if models_parameters['stateful']:
            maximum_sequences = X_fault.shape[0] - X_fault.shape[0] % models_parameters['batch_size']
            X_fault = X_fault[:maximum_sequences]
            y_fault = y_fault[:maximum_sequences]
        
        if measure_time:
            start_time_predict = time.perf_counter()

        y_hat_fault = model.predict(X_fault, batch_size = models_parameters['batch_size'], verbose = 0)
        
        if measure_time:
            end_time_predict = time.perf_counter()
            results_time_measure_fault_data.append(end_time_predict - start_time_predict)


        y_f = y_fault.flatten()
        y_hf = y_hat_fault.flatten()

        # Get residuals
        residuals_fault = (y_f - y_hf)**2

    results_time_measure['test'] = results_time_measure_fault_data

    return residuals_clean, residuals_fault,  results_time_measure
````

````python
# Estimate CUSUMs parameters function
def estimate_cusums_parameters(param_estimation,
                               classic_threshold = False,
                               cusums_parameters = None,
                               max_threshold = False):
    
    """ 
        Estimate CUSUMs parameters
        Arguments:
        ---------
        param_estimation: Data used to estimate the parameters
        classic_threshold: True to use the classic threshold
        cusums_parameters: Dictionary with CUSUMs parameters
        
        Returns:
        --------
        estimated_hs: Estimated thresholds
        estimated_ks: Estimated reference parameters
        target_mean_std: Target mean and std for either data or CUSUM
    """
    
    if cusums_parameters is None:
        raise ValueError("cusums_parameters is None")

    estimated_hs = {}
    estimated_ks = {}
    target_mean_std = {}    

    # Mean and STD of the Param Estimation Data
    target_mean = estimate_target_mean(param_estimation)
    target_std = np.std(param_estimation)
    
    target_mean_std['all'] = (target_mean, target_std)
    
    # High Order CUSUMs
    exp_x, k_estimated_ho = np.abs(estimate_k_from_stats(data = param_estimation, 
                                                         shift_size=cusums_parameters['high_order']['shift_size'],
                                                         alpha=cusums_parameters['high_order']['alpha']))
    

    if not classic_threshold:
        cusum_stat_pos, cusum_stat_neg, _, _ = high_order_cusum(param_estimation, 
                                            window = cusums_parameters['high_order']['window'], 
                                            alpha = cusums_parameters['high_order']['alpha'],
                                            exp_x=exp_x,
                                            k=k_estimated_ho)

        h_estimated_ho = estimate_cusum_threshold(cusum_stats_pos=cusum_stat_pos, 
                                                cusum_stats_neg=cusum_stat_neg,
                                                clean_data=None, 
                                                h=cusums_parameters['high_order']['h'], 
                                                csm_type="high_order",
                                                window=cusums_parameters['high_order']['window'],
                                                alpha=cusums_parameters['high_order']['alpha'],
                                                max_threshold=max_threshold)
    
    else:
        h_estimated_ho = estimate_cusum_threshold(clean_data=param_estimation, 
                            h = cusums_parameters['high_order']['h'],
                            max_threshold=max_threshold)

    estimated_hs['high_order'] = h_estimated_ho
    estimated_ks['high_order'] = k_estimated_ho
    target_mean_std['high_order'] = exp_x

    # Classical CUSUMs
    k_estimated_classical = estimate_k_from_shift(std_dev=target_std, 
                                                  shift_size=cusums_parameters['classical']['shift_size'])

    if classic_threshold:
        h_estimated_classical = estimate_cusum_threshold(clean_data=param_estimation, 
                                        h= cusums_parameters['classical']['h'],
                                        max_threshold=max_threshold)
    else:
        pos_clean, neg_clean, _ = classical_cusum(param_estimation, 
                                        target_mean=target_mean, 
                                        k=k_estimated_classical, 
                                        h=0)
    
        pos_clean = np.maximum(abs(pos_clean), abs(neg_clean))

        h_estimated_classical = estimate_cusum_threshold(clean_data=pos_clean, 
                                        h = cusums_parameters['classical']['h'],
                                        max_threshold=max_threshold)
    

    estimated_hs['classical'] = h_estimated_classical
    estimated_ks['classical'] = k_estimated_classical


    # One CUSUM Scheme
    k_estimated_one = estimate_k_from_shift(std_dev=target_std, 
                                            shift_size=cusums_parameters['one_cusum_scheme']['shift_size'])

    if classic_threshold:
        h_estimated_one = estimate_cusum_threshold(clean_data=param_estimation, 
                                        h=cusums_parameters['one_cusum_scheme']['h'],
                                        max_threshold=max_threshold)
    else:
        cusum_stat_clean, _ = one_cusum_scheme(param_estimation, 
                                        target_mean=target_mean, 
                                        k=k_estimated_one, 
                                        h=0, 
                                        lamb=cusums_parameters['one_cusum_scheme']['lamb']) 

        h_estimated_one = estimate_cusum_threshold(clean_data=cusum_stat_clean, 
                                        h=cusums_parameters['one_cusum_scheme']['h'],
                                        max_threshold=max_threshold)
    

    estimated_hs['one_cusum_scheme'] = h_estimated_one
    estimated_ks['one_cusum_scheme'] = k_estimated_one

    # Hampel CUSUMs
    k_estimated_hampel = estimate_k_from_shift(std_dev=target_std, 
                                               shift_size=cusums_parameters['hampel_acusum']['shift_size'])

    if classic_threshold:
        h_estimated_hampel = estimate_cusum_threshold(clean_data=param_estimation, 
                                        h=cusums_parameters['hampel_acusum']['h'],
                                        csm_type="hampel_acusum",
                                        max_threshold=max_threshold)
    else:
        cusum_stat_pos_clean, cusum_stat_neg_clean, _ = \
                                    hampel_acusum(param_estimation, 
                                       a=cusums_parameters['hampel_acusum']['a'], 
                                       b=cusums_parameters['hampel_acusum']['b'], 
                                       r=cusums_parameters['hampel_acusum']['r'], 
                                       gamma=cusums_parameters['hampel_acusum']['gamma'], 
                                       h=0, 
                                       k_plus=k_estimated_hampel, 
                                       k_minus=k_estimated_hampel)
        
        cusum_stat_pos_clean = np.maximum(abs(cusum_stat_pos_clean), abs(cusum_stat_neg_clean))
        
        h_estimated_hampel = estimate_cusum_threshold(clean_data=cusum_stat_pos_clean, 
                                        h=cusums_parameters['hampel_acusum']['h'],
                                        csm_type="hampel_acusum",
                                        max_threshold=max_threshold)



    estimated_hs['hampel_acusum'] = h_estimated_hampel
    estimated_ks['hampel_acusum'] = k_estimated_hampel

    # UCUSUM
    if classic_threshold:
        h_estimated_ucusum = estimate_cusum_threshold(clean_data=param_estimation, 
                                                    h=cusums_parameters['UCUSUM']['h'],
                                                    csm_type="UCUSUM",
                                                    max_threshold=max_threshold)  
    else:
        cusum_stat_clean, _ = UCUSUM(param_estimation, 
                                     window=cusums_parameters['UCUSUM']['window'],
                                    target_mean=target_mean, 
                                    h=0)
        h_estimated_ucusum = estimate_cusum_threshold(clean_data=cusum_stat_clean, 
                                                    h=cusums_parameters['UCUSUM']['h'],
                                                    csm_type="UCUSUM",
                                                    max_threshold=max_threshold)  


    estimated_hs['UCUSUM'] = h_estimated_ucusum

    # Variance CUSUM
    k_estimated_variance = estimate_k_from_shift(std_dev=target_std, 
                                                 shift_size=cusums_parameters['variance_cusum']['shift_size'])
    
    if classic_threshold:
        h_estimated_variance = estimate_cusum_threshold(clean_data=param_estimation, 
                                            h=cusums_parameters['variance_cusum']['h'],
                                            csm_type="variance_cusum",
                                            max_threshold=max_threshold)
    else:
        cusum_stat_clean_pos, cusum_stat_clean_neg, _ = variance_cusum(param_estimation, 
                                            target_std=target_std, 
                                            window=cusums_parameters['variance_cusum']['window'], 
                                            k=k_estimated_variance,
                                            h=0)
        cusum_stat_clean_pos = np.maximum(abs(cusum_stat_clean_pos), abs(cusum_stat_clean_neg))
        
        h_estimated_variance = estimate_cusum_threshold(clean_data=cusum_stat_clean_pos, 
                                            h=cusums_parameters['variance_cusum']['h'],
                                            csm_type="variance_cusum",
                                            max_threshold=max_threshold)
        

    estimated_hs['variance_cusum'] = h_estimated_variance
    estimated_ks['variance_cusum'] = k_estimated_variance

    return estimated_hs, estimated_ks, target_mean_std
````

````python
# Function to run all CUSUMs
def run_all_cusums2(clean = None, 
                    fault = None,
                    labels_clean = None,
                    labels_fault = None,
                    dataset="", 
                    datasets_parameters = {},
                    cusums_parameters = {},
                    run_on_residuals = False,
                    classic_threshold = False,
                    max_threshold=False):
    """
        Function to run all CUSUMs
        - Set the parameters in the cusums_parameters dictionary
        - Set the parameters in the datasets_parameters dictionary
        - Set the parameters in the models_parameters dictionary
        - Feel free to add remove datasets, CUSUMs and parameters
        
        Arguments:
        ----------
            clean: clean data np array
            fault: fault data np array or list of np arrays
            labels_clean: clean labels np array
            labels_fault: fault labels np array
            dataset: dataset name
            datasets_parameters: datasets parameters dictionary
            run_on_residuals: True to run the CUSUMs on the residuals
            cusums_parameters: CUSUMs parameters dictionary
            classic_threshold: True to use the classic threshold
            max_threshold: Max threshold for CUSUMs
        Returns:
            cusum_detection_results: CUSUMs detection results dict
                contains: labels, detections on clean and fault data
            dataset: dataset name
    """
    # Params.
    param_estimation_size = datasets_parameters['param_estimation_size']
    variable = datasets_parameters['variable']
    simulation = datasets_parameters['simulation']
    sim_length = datasets_parameters['sim_length'] if datasets_parameters['simulation'] is True else 0
    clean_sim_length = datasets_parameters['clean_sim_length'] if datasets_parameters['simulation'] is True else 0


    param_size = int(len(clean) * param_estimation_size)
    param_estimation = np.array(clean[:param_size])
    clean = np.array(clean[param_size:])


    #######################################################################################
    # Estimate Parameters for CUSUMs
    
    estimated_hs, estimated_ks, target_mean_std = \
                    estimate_cusums_parameters(param_estimation,
                                            classic_threshold = classic_threshold,
                                            cusums_parameters = cusums_parameters,
                                            max_threshold=max_threshold)
    
    target_mean = target_mean_std['all'][0]
    target_std = target_mean_std['all'][1]

    #######################################################################################
    # CUSUMS Detections on Clean Data and Faulty Data

    # Detection Results    
    cusum_detection_results = { }
    
    # High Order CUSUMs
    h_estimated_ho = estimated_hs['high_order']
    k_estimated_ho = estimated_ks['high_order']
    exp_x = target_mean_std['high_order']

    _, _, _, detections_clean = high_order_cusum(clean, 
                                            window = cusums_parameters['high_order']['window'], 
                                            alpha = cusums_parameters['high_order']['alpha'],
                                            exp_x=exp_x,
                                            k=k_estimated_ho, 
                                            h=h_estimated_ho,
                                            simulation=False,
                                            sim_length=clean_sim_length)
    
    _, _, _, detections_fault = high_order_cusum(fault, 
                                            window = cusums_parameters['high_order']['window'],
                                            alpha = cusums_parameters['high_order']['alpha'],
                                            exp_x=exp_x,
                                            k=k_estimated_ho, 
                                            h=h_estimated_ho,
                                            simulation=simulation,
                                            sim_length=sim_length
                                            )
        

    cusum_detection_results['high_order'] = {
                        'labels_clean': labels_clean,
                        'detections_clean': detections_clean,
                        'labels_fault': labels_fault,
                        'detections_fault': detections_fault,
                        'threshold': h_estimated_ho
                    }

    # cleanup
    gc.collect()
    del detections_clean, detections_fault


    # Classical CUSUMs
    h_estimated_classical = estimated_hs['classical']
    k_estimated_classical = estimated_ks['classical']

    _, _, detections_clean = classical_cusum(clean, 
                                       target_mean=target_mean, 
                                       k=k_estimated_classical, 
                                       h=h_estimated_classical,
                                       simulation=False,
                                       sim_length=clean_sim_length)

    _, _, detections_fault = classical_cusum(fault, 
                                            target_mean=target_mean, 
                                            k=k_estimated_classical, 
                                            h=h_estimated_classical,
                                            simulation=simulation,
                                            sim_length=sim_length)
        

    cusum_detection_results['classical'] = {
                        'labels_clean': labels_clean,
                        'detections_clean': detections_clean,
                        'labels_fault': labels_fault,
                        'detections_fault': detections_fault,
                        'threshold': h_estimated_classical
                    }

    # cleanup
    gc.collect()
    del detections_clean, detections_fault

    # One CUSUM Scheme
    h_estimated_one = estimated_hs['one_cusum_scheme']
    k_estimated_one = estimated_ks['one_cusum_scheme']

    _, detections_clean = one_cusum_scheme(clean, 
                                          target_mean=target_mean, 
                                          k=k_estimated_one, 
                                          h=h_estimated_one, 
                                          lamb=cusums_parameters['one_cusum_scheme']['lamb'], 
                                          simulation=False, 
                                          sim_length=clean_sim_length) 

    _, detections_fault = one_cusum_scheme(fault, 
                                            target_mean=target_mean, 
                                            k=k_estimated_one, 
                                            h=h_estimated_one, 
                                            lamb=cusums_parameters['one_cusum_scheme']['lamb'],
                                            simulation=simulation,
                                            sim_length=sim_length) 
        

    cusum_detection_results['one_cusum_scheme'] = {
                        'labels_clean': labels_clean,
                        'detections_clean': detections_clean,
                        'labels_fault': labels_fault,
                        'detections_fault': detections_fault,
                        'threshold': h_estimated_one
                    }

    # cleanup
    gc.collect()
    del  detections_clean,  detections_fault

    # Hampel CUSUMs
    h_estimated_hampel = estimated_hs['hampel_acusum']
    k_estimated_hampel = estimated_ks['hampel_acusum']


    _, _, detections_clean = \
                                    hampel_acusum(clean, 
                                       a=cusums_parameters['hampel_acusum']['a'], 
                                       b=cusums_parameters['hampel_acusum']['b'], 
                                       r=cusums_parameters['hampel_acusum']['r'], 
                                       gamma=cusums_parameters['hampel_acusum']['gamma'], 
                                       h=h_estimated_hampel, 
                                       k_plus=k_estimated_hampel, 
                                       k_minus=k_estimated_hampel,
                                       simulation=False,
                                       sim_length=clean_sim_length)

    _, _, detections_fault = \
                                    hampel_acusum(fault, 
                                        a=cusums_parameters['hampel_acusum']['a'], 
                                        b=cusums_parameters['hampel_acusum']['b'], 
                                        r=cusums_parameters['hampel_acusum']['r'], 
                                        gamma=cusums_parameters['hampel_acusum']['gamma'], 
                                        h=h_estimated_hampel, 
                                        k_plus=k_estimated_hampel, 
                                        k_minus=k_estimated_hampel,
                                        simulation=simulation,
                                        sim_length=sim_length)
        

    cusum_detection_results['hampel'] = {
                        'labels_clean': labels_clean,
                        'detections_clean': detections_clean,
                        'labels_fault': labels_fault,
                        'detections_fault': detections_fault,
                        'threshold': h_estimated_hampel}
    # cleanup
    gc.collect()
    del detections_clean, detections_fault

    # UCUSUM Mean Shift

    h_estimated_ucusum = estimated_hs['UCUSUM']

    _, detections_clean = UCUSUM(clean, 
                                        target_mean=target_mean, 
                                        h=h_estimated_ucusum,
                                        simulation=False,
                                        sim_length=clean_sim_length)
    

    _, detections_fault = UCUSUM(fault,
                                        target_mean=target_mean, 
                                        h=h_estimated_ucusum,
                                        simulation=simulation,
                                        sim_length=sim_length)
    

    cusum_detection_results['UCUSUM'] = {
                        'labels_clean': labels_clean,
                        'detections_clean': detections_clean,
                        'labels_fault': labels_fault,
                        'detections_fault': detections_fault,
                        'threshold': h_estimated_ucusum
                    }

    # cleanup
    gc.collect()
    del detections_clean, detections_fault

    # Variance CUSUM
    h_estimated_variance = estimated_hs['variance_cusum']
    k_estimated_variance = estimated_ks['variance_cusum']

    _, _, detections_clean = variance_cusum(clean, 
                                        target_std=target_std, 
                                        window=cusums_parameters['variance_cusum']['window'], 
                                        k=k_estimated_variance, 
                                        h=h_estimated_variance,
                                        simulation=False,
                                        sim_length=clean_sim_length)


    _, _, detections_fault = variance_cusum(fault, 
                                        target_std=target_std, 
                                        window=cusums_parameters['variance_cusum']['window'], 
                                        k=k_estimated_variance, 
                                        h=h_estimated_variance,
                                        simulation=simulation,
                                        sim_length=sim_length)
        

    cusum_detection_results['variance_cusum'] = {
                        'labels_clean': labels_clean,
                        'detections_clean': detections_clean,
                        'labels_fault': labels_fault,
                        'detections_fault': detections_fault, 
                        'threshold': h_estimated_variance
                    }
    # cleanup
    del  detections_clean, detections_fault

    return cusum_detection_results, dataset
````

````python
# Function to run all CUSUMs on clean with noise.
def run_cusums_on_clean_with_noise(
                    clean = None, 
                    fault = None,
                    labels_clean = None,
                    run_on_residuals = False,
                    datasets_parameters = {},
                    cusums_parameters = {},
                    classic_threshold = False,
                    max_threshold=False,
                    noise = [0, 0, 0],
                    outliers = [0, 0, 0],
                    add_to_param = False):
    """
    Function to run all CUSUMs on clean with noise and injected outliers.
    - Set the parameters in the cusums_parameters dictionary
    - Set the parameters in the datasets_parameters dictionary
    - Set the parameters in the models_parameters dictionary
    - Feel free to add remove datasets, CUSUMs and parameters
    Arguments:
        clean: clean data np array
        fault: fault data np array or list of np arrays
        labels_clean: clean labels np array
        datasets_parameters: datasets parameters dictionary
        cusums_parameters: CUSUMs parameters dictionary
        run_on_residuals: True to run the CUSUMs on the residuals
        classic_threshold: True to use the classic threshold
        max_threshold: Max threshold for CUSUMs
        noise: list of 3 elements [type, mean (or min), std (or max)]
            type: 0 - normal, 1 - uniform
            mean (or min): mean for normal, min for uniform
            std (or max): std for normal, max for uniform
        outliers: list of 3 elements [percentage, min, max]
            percentage: percentage of outliers to add to the clean data
            min: minimum value of the outliers (as a percentage of the data min)
            max: maximum value of the outliers (as a percentage of the data max)
        add_to_param: True to add noise and outliers to the parameter estimation data
    Returns:
        cusum_detection_results: CUSUMs detection results dict
            contains: labels, detections on clean data
    """
    
    # Params.
    param_estimation_size = datasets_parameters['param_estimation_size']
    variable = datasets_parameters['variable']

    simulation = datasets_parameters['simulation']
    clean_sim_length = datasets_parameters['clean_sim_length'] if datasets_parameters['simulation'] is True else 0


    param_size = int(len(clean) * param_estimation_size)
    param_estimation = np.array(clean[:param_size])
    clean = np.array(clean[param_size:])

    # ADD NOISE to the clean data
    # Noise = [0 - normal, 1 - uniform, Mean (or MIN), STD or (MAX)]
    # As data is not normalized for the raw signals verios, 
    # the noise is added as percentage not as absolute values (RELATIVE NOISE!)

    noise_mean_or_min = noise[1]
    noise_std_or_max = noise[2] * np.std(clean)

    if noise != [0, 0, 0] and noise != [1, 0, 0]:

        if noise[0] == 0:
            gen_noise = np.random.normal(noise_mean_or_min, noise_std_or_max, size = clean.shape)
        elif noise[0] == 1:
            gen_noise = np.random.uniform(noise_mean_or_min, noise_std_or_max, size = clean.shape)
            
        clean = clean + gen_noise

        if add_to_param:
            if noise[0] == 0:
                gen_noise_param = np.random.normal(noise_mean_or_min, noise_std_or_max, size = param_estimation.shape)
            elif noise[0] == 1:
                gen_noise_param = np.random.uniform(noise_mean_or_min, noise_std_or_max, size = param_estimation.shape)
            
            param_estimation = param_estimation + gen_noise_param

    # Randomly add a percentage of outliers to the clean data
    # The percentage is given by outliers variable, at random positions.
    # Outliers are largers than the data itself

    if outliers != [0, 0, 0]:
        outliers_percentage_clean = int(outliers[0] * clean.shape[0])

        outliers_min = outliers[1] * clean.min()
        outliers_max = outliers[2] * clean.max()

        outliers_positions = np.random.choice(clean.shape[0], size = outliers_percentage_clean, replace = False)
        outliers_positions = np.sort(outliers_positions)


        if add_to_param:
            outliers_percentage_param = int(outliers[0] * param_estimation.shape[0])

            outliers_min = outliers[1] * param_estimation.min()
            outliers_max = outliers[2] * param_estimation.max()

            outliers_positions_param = np.random.choice(param_estimation.shape[0], size = outliers_percentage_param, replace = False)
            outliers_positions_param = np.sort(outliers_positions_param)

        # Random outliers, positive or negative
        for idx in outliers_positions:
            #sign = np.sign(clean[idx])
            sign = np.random.choice([-1, 1])
            magnitude = np.random.uniform(outliers_min, outliers_max)
            clean[idx] = sign * magnitude 

        if add_to_param:
            for idx in outliers_positions_param:
                #sign = np.sign(param_estimation[idx])
                sign = np.random.choice([-1, 1])
                magnitude = np.random.uniform(outliers_min, outliers_max)
                param_estimation[idx] = sign * magnitude

   #######################################################################################
    # Estimate Parameters for CUSUMs

    estimated_hs, estimated_ks, target_mean_std = \
                    estimate_cusums_parameters(param_estimation,
                                            classic_threshold = classic_threshold,
                                            cusums_parameters = cusums_parameters,
                                            max_threshold = max_threshold)
    
    target_mean = target_mean_std['all'][0]
    target_std = target_mean_std['all'][1]

    #######################################################################################

    # Detection Results    
    cusum_detection_results = { }
    # CUSUMS Detections on Clean Noisy Data

    # High Order CUSUMs
    h_estimated_ho = estimated_hs['high_order']
    k_estimated_ho = estimated_ks['high_order']
    exp_x = target_mean_std['high_order']

    _, _, _, detections_clean = high_order_cusum(clean, 
                                            window = cusums_parameters['high_order']['window'], 
                                            alpha = cusums_parameters['high_order']['alpha'],
                                            exp_x=exp_x,
                                            k=k_estimated_ho, 
                                            h=h_estimated_ho,
                                            simulation=False,
                                            sim_length=clean_sim_length)
    
    cusum_detection_results['high_order'] = {
                        'labels_clean': labels_clean,
                        'detections_clean': detections_clean,
                        'threshold': h_estimated_ho}
    # cleanup
    gc.collect()
    del detections_clean

    # Classical CUSUMs
    h_estimated_classical = estimated_hs['classical']
    k_estimated_classical = estimated_ks['classical']

    _, _, detections_clean = classical_cusum(clean, 
                                       target_mean=target_mean, 
                                       k=k_estimated_classical, 
                                       h=h_estimated_classical, 
                                       simulation=False,
                                       sim_length=clean_sim_length)
    
    cusum_detection_results['classical'] = {
                        'labels_clean': labels_clean,
                        'detections_clean': detections_clean,
                        'threshold': h_estimated_classical}
    # cleanup
    gc.collect()
    del detections_clean

    # One CUSUM Scheme
    h_estimated_one = estimated_hs['one_cusum_scheme']
    k_estimated_one = estimated_ks['one_cusum_scheme']

    _, detections_clean = one_cusum_scheme(clean, 
                                          target_mean=target_mean, 
                                          k=k_estimated_one, 
                                          h=h_estimated_one, 
                                          lamb=cusums_parameters['one_cusum_scheme']['lamb'],
                                          simulation=False, 
                                          sim_length=clean_sim_length) 

    cusum_detection_results['one_cusum_scheme'] = {
                        'labels_clean': labels_clean,
                        'detections_clean': detections_clean,
                        'threshold': h_estimated_one}
    
    # cleanup
    gc.collect()
    del detections_clean

    # Hampel CUSUMs
    h_estimated_hampel = estimated_hs['hampel_acusum']
    k_estimated_hampel = estimated_ks['hampel_acusum']
    _, _, detections_clean = \
                                    hampel_acusum(clean, 
                                       a=cusums_parameters['hampel_acusum']['a'], 
                                       b=cusums_parameters['hampel_acusum']['b'], 
                                       r=cusums_parameters['hampel_acusum']['r'], 
                                       gamma=cusums_parameters['hampel_acusum']['gamma'], 
                                       h=h_estimated_hampel, 
                                       k_plus=k_estimated_hampel, 
                                       k_minus=k_estimated_hampel,
                                       simulation=False,
                                       sim_length=clean_sim_length)

    cusum_detection_results['hampel'] = {
                        'labels_clean': labels_clean,
                        'detections_clean': detections_clean,
                        'threshold': h_estimated_hampel}
    
    gc.collect()
    del detections_clean

    # UCUSUM
    h_estimated_ucusum = estimated_hs['UCUSUM']

    _, detections_clean = UCUSUM(clean, 
                                        target_mean=target_mean, 
                                        h=h_estimated_ucusum)
    
    cusum_detection_results['UCUSUM'] = {
                        'labels_clean': labels_clean,
                        'detections_clean': detections_clean,
                        'threshold': h_estimated_ucusum}
    gc.collect()
    del detections_clean

    # Variance CUSUM
    h_estimated_variance = estimated_hs['variance_cusum']
    k_estimated_variance = estimated_ks['variance_cusum']
    
    _, _, detections_clean = variance_cusum(clean, 
                                        target_std=target_std, 
                                        window=cusums_parameters['variance_cusum']['window'], 
                                        k=k_estimated_variance, 
                                        h=h_estimated_variance, 
                                        simulation=False,
                                        sim_length=clean_sim_length)
    
    cusum_detection_results['variance_cusum'] = {
                        'labels_clean': labels_clean,
                        'detections_clean': detections_clean,
                        'threshold': h_estimated_variance}
    
    return cusum_detection_results
````

````python
# Function to run all CUSUMs on fault for resilience
def run_cusums_for_resilience(
                    dataset = 'TEP',
                    clean = None, 
                    fault = None,
                    labels_fault = None,
                    datasets_parameters = {},
                    cusums_parameters = {},
                    run_on_residuals = False,
                    classic_threshold = False,
                    max_threshold = False):
    """
    Function to run all CUSUMs on clean with noise and injected outliers.
    - Set the parameters in the cusums_parameters dictionary
    - Set the parameters in the datasets_parameters dictionary
    - Set the parameters in the models_parameters dictionary
    - Feel free to add remove datasets, CUSUMs and parameters
    Parameters
    ----------
    clean : np.array
        Clean data array
    fault : np.array or list of np.array
        Faulty data array or list of arrays
    datasets_parameters : dict
        Dictionary with the datasets parameters
    cusums_parameters : dict
        Dictionary with the CUSUMs parameters
    run_on_residuals : bool, optional
        True to run the CUSUMs on the residuals, by default False
    classic_threshold : bool, optional
        True to use the classic threshold, by default False
    max_threshold : bool, optional
        Max threshold for CUSUMs, by default False
    labels_fault : np.array, optional
        Fault labels array, by default None
    Returns
    -------
    cusum_detection_results : dict
        Dictionary with the CUSUM detection results
    """

    param_estimation_size = datasets_parameters['param_estimation_size']
    variable = datasets_parameters['variable']
    max_length = datasets_parameters['max_length']
    simulation = datasets_parameters['simulation']
    sim_length = datasets_parameters['sim_length'] if datasets_parameters['simulation'] is True else 0
    clean_sim_length = datasets_parameters['clean_sim_length']
    
    # Get Clean and Fault Data
    residuals_text = "residuals_" if run_on_residuals else ""

    resilience_data_file = "datasets/resilience_data_" + \
                        str(max_length) + "_" + \
                        str(param_estimation_size) + "_" +\
                        str(variable) + "_" + \
                        str(residuals_text) + "_" + \
                        dataset + ".pkl"
    
    # If any data is saved, load it
    if os.path.exists(resilience_data_file):
        resilience_data = load_pickle(resilience_data_file)

        param_estimation = resilience_data['param_estimation']
        fault = resilience_data['fault']
        labels_fault = resilience_data['labels_fault']
        datasets_parameters['resilience_sim_length'] = resilience_data['resilience_sim_length']

    else:

        param_size = int(len(clean) * param_estimation_size)
        param_estimation = np.array(clean[:param_size])
        clean = np.array(clean[param_size:])
    
        # Resilience experiment. As such, we will concatenate the clean data afetr the faulty data.
        RESILIENCE_CLEAN_DATA_MULTIPLIER = datasets_parameters['RESILIENCE_CLEAN_DATA_MULTIPLIER']

        if simulation == True:
            if not isinstance(sim_length, list):
                clean_tmp = np.array(clean[0: RESILIENCE_CLEAN_DATA_MULTIPLIER * clean_sim_length], dtype=float)
                max_rows = fault.shape[0] - fault.shape[0] % sim_length
                num_simulations = fault.shape[0] // sim_length
                
                clean_rows = np.tile(clean_tmp, (num_simulations, 1))

                fault_reshaped = fault[0:max_rows].reshape((num_simulations, sim_length))
                combined = np.concatenate((fault_reshaped, clean_rows), axis=1)
                datasets_parameters['resilience_sim_length'] = combined.shape[1]
                fault = combined.reshape(-1)


                # Same magic for labels
                labels_clean_tmp = np.zeros(len(clean_rows[0]))
                
                labels_fault_reshaped = labels_fault[0:max_rows].reshape((num_simulations, sim_length))
                labels_clean_rows = np.tile(labels_clean_tmp, (num_simulations, 1))
                
                combined = np.concatenate((labels_fault_reshaped, labels_clean_rows), axis=1)
                labels_fault = combined.reshape(-1)


            else:
                # Different simulation lengths
                clean_tmp = np.array(clean[0: (RESILIENCE_CLEAN_DATA_MULTIPLIER + 3) * clean_sim_length], dtype=float)
                
                largest_sim_length = max(sim_length)
                clean_differences = [abs(largest_sim_length - s) for s in sim_length]
                num_simulations = len(sim_length)
                max_rows = num_simulations
                
                fault_data_conc = []
                current_index = 0 # CUMULATE SIM LENGTHS
                addition = RESILIENCE_CLEAN_DATA_MULTIPLIER * clean_sim_length

                for i in range(num_simulations):

                    clean_row = clean_tmp[:addition + clean_differences[i]]

                    sl = sim_length[i]
                    
                    fault_data_sim = fault[current_index:current_index+sl]
                    
                    if len(fault_data_sim) < sl:
                        clean_row = clean_tmp[:addition + clean_differences[i] + sl - len(fault_data_sim)]

                    fault_data_conc.append(np.concatenate((fault_data_sim, clean_row), axis=0))
                    
                    current_index += sl
        
                datasets_parameters['resilience_sim_length'] = fault_data_conc[0].shape[0]
                fault = np.array(fault_data_conc).reshape(-1)
                
                # Same magic for labels
                labels_clean_tmp = np.zeros((RESILIENCE_CLEAN_DATA_MULTIPLIER + 3) * clean_sim_length)
                labels_fault_data_conc = []
                addition = RESILIENCE_CLEAN_DATA_MULTIPLIER * clean_sim_length
                current_index = 0 

                for i in range(num_simulations):
                    clean_row = labels_clean_tmp[:addition + clean_differences[i]]

                    sl = sim_length[i]

                    labels_fault_sim = labels_fault[current_index:current_index+sl]

                    if len(labels_fault_sim) < sl:
                        clean_row = labels_clean_tmp[:addition + clean_differences[i] + sl - len(labels_fault_sim)]

                    labels_fault_data_conc.append(np.concatenate((labels_fault_sim, clean_row), axis=0))
                    current_index += sl

                labels_fault = np.array(labels_fault_data_conc).reshape(-1)
                
        resilience_data = {
            "param_estimation": param_estimation,
            "fault": fault,
            "labels_fault": labels_fault,
            "resilience_sim_length": datasets_parameters['resilience_sim_length']
        }

        save_pickle(resilience_data, resilience_data_file)

    # Data structures used from this point on:
    # param_estimation, fault, labels_clean, labels_fault

    sim_length = datasets_parameters['resilience_sim_length']

   #######################################################################################
    # Estimate Parameters for CUSUMs

    estimated_hs, estimated_ks, target_mean_std = \
                    estimate_cusums_parameters(param_estimation,
                                            classic_threshold = classic_threshold,
                                            cusums_parameters = cusums_parameters,
                                            max_threshold=max_threshold)
    
    target_mean = target_mean_std['all'][0]
    target_std = target_mean_std['all'][1]

    #######################################################################################
    # Detection Results    
    cusum_detection_results = { }

    # CUSUMS Detections on Clean Data and Faulty Data
    
    # High Order CUSUMs
    h_estimated_ho = estimated_hs['high_order']
    k_estimated_ho = estimated_ks['high_order']
    exp_x = target_mean_std['high_order']

    cusum_stat_pos, cusum_stat_neg, _, detections_fault = high_order_cusum(fault, 
                                            window = cusums_parameters['high_order']['window'],
                                            alpha = cusums_parameters['high_order']['alpha'],
                                            exp_x=exp_x,
                                            k=k_estimated_ho, 
                                            h=h_estimated_ho,
                                            simulation=simulation,
                                            sim_length=sim_length,
                                            resilience=False)

    cusum_detection_results['high_order'] = {
                        'labels_fault': labels_fault,
                        'cusum_stat_pos': cusum_stat_pos,
                        'cusum_stat_neg': cusum_stat_neg,
                        'detections_fault': detections_fault,
                        'threshold': h_estimated_ho}

    # cleanup
    gc.collect()
    del detections_fault

    # Classical CUSUMs
    h_estimated_classical = estimated_hs['classical']
    k_estimated_classical = estimated_ks['classical']


    pos_fault, neg_fault, detections_fault = classical_cusum(fault, 
                                            target_mean=target_mean, 
                                            k=k_estimated_classical, 
                                            h=h_estimated_classical,
                                            simulation=simulation,
                                            sim_length=sim_length,
                                            resilience=False)
    
    cusum_detection_results['classical'] = {
                        'labels_fault': labels_fault,
                        'cusum_stat_pos': pos_fault,
                        'cusum_stat_neg': neg_fault,
                        'detections_fault': detections_fault,
                        'threshold': h_estimated_classical
                    }

    # cleanup
    gc.collect()
    del detections_fault

    # One CUSUM Scheme
    h_estimated_one = estimated_hs['one_cusum_scheme']
    k_estimated_one = estimated_ks['one_cusum_scheme']


    cusum_stat_fault, detections_fault = one_cusum_scheme(fault, 
                                            target_mean=target_mean, 
                                            k=k_estimated_one, 
                                            h=h_estimated_one, 
                                            lamb=cusums_parameters['one_cusum_scheme']['lamb'],
                                            simulation=simulation,
                                            sim_length=sim_length,
                                            resilience=False) 
    
    cusum_detection_results['one_cusum_scheme'] = {
                        'labels_fault': labels_fault,
                        'cusum_stat_pos': cusum_stat_fault,
                        'cusum_stat_neg': None,
                        'detections_fault': detections_fault,
                        'threshold': h_estimated_one
                    }

    # cleanup
    gc.collect()
    del  detections_fault

    # Hampel CUSUMs
    h_estimated_hampel = estimated_hs['hampel_acusum']
    k_estimated_hampel = estimated_ks['hampel_acusum']


    cusum_stat_pos_fault, cusum_stat_neg_fault, detections_fault = \
                                    hampel_acusum(fault, 
                                        a=cusums_parameters['hampel_acusum']['a'], 
                                        b=cusums_parameters['hampel_acusum']['b'], 
                                        r=cusums_parameters['hampel_acusum']['r'], 
                                        gamma=cusums_parameters['hampel_acusum']['gamma'], 
                                        h=h_estimated_hampel, 
                                        k_plus=k_estimated_hampel, 
                                        k_minus=k_estimated_hampel,
                                        simulation=simulation,
                                        sim_length=sim_length,
                                        resilience=False)
    
    cusum_detection_results['hampel'] = {
                        'labels_fault': labels_fault,
                        'cusum_stat_pos': cusum_stat_pos_fault,
                        'cusum_stat_neg': cusum_stat_neg_fault,
                        'detections_fault': detections_fault,
                        'threshold': h_estimated_hampel}
    # cleanup
    gc.collect()
    del detections_fault


    # UCUSUM Mean Shift
    
    h_estimated_ucusum = estimated_hs['UCUSUM']


    cusum_stat_fault, detections_fault = UCUSUM(fault,
                                        target_mean=target_mean, 
                                        h=h_estimated_ucusum,
                                        simulation=simulation,
                                        sim_length=sim_length,
                                        resilience=False)
    

    cusum_detection_results['UCUSUM'] = {
                        'labels_fault': labels_fault,
                        'cusum_stat_pos': cusum_stat_fault,
                        'cusum_stat_neg': None,
                        'detections_fault': detections_fault,
                        'threshold': h_estimated_ucusum
                    }

    # cleanup
    gc.collect()
    del detections_fault

    # Variance CUSUM
    h_estimated_variance = estimated_hs['variance_cusum']
    k_estimated_variance = estimated_ks['variance_cusum']

    cusum_stat_pos_fault, cusum_stat_neg_fault, detections_fault = variance_cusum(fault, 
                                        target_std=target_std, 
                                        window=cusums_parameters['variance_cusum']['window'], 
                                        k=k_estimated_variance, 
                                        h=h_estimated_variance,
                                        simulation=simulation,
                                        sim_length=sim_length,
                                        resilience=False)
    

    cusum_detection_results['variance_cusum'] = {
                        'labels_fault': labels_fault,
                        'cusum_stat_pos': cusum_stat_pos_fault,
                        'cusum_stat_neg': cusum_stat_neg_fault,
                        'detections_fault': detections_fault, 
                        'threshold': h_estimated_variance
                    }
    # cleanup
    del detections_fault

    return cusum_detection_results
````

````python
# Function to compute all results

def compute_results(cusum_detection_results = None, 
                    dataset = "TEP", 
                    mode = "standard",
                    simulation = False,
                    sim_length = 1000,
                    logger = None):
    """
    Compute all results for all CUSUMs

    Parameters
    ----------
    cusum_detection_results : list or np.array
        Data structure with all CUSUM detection results
    dataset : str
        Dataset name
    mode : str
        Mode of the experiment
            - standard: FPR, TPR, DD
            - robust or resilience: robustness or resilience
    """

    if cusum_detection_results is None:
        raise ValueError("CUSUM detection results is None")

    if mode not in ["standard", "robust", "resilience"]:
        raise ValueError(f"Mode {mode} is not supported \
                         - supported modes are: \
                         'standard', 'robust', 'resilience'")

    if mode == 'standard':
        # Compute FPR and TPR for each CUSUM method
        for csm, detection in cusum_detection_results.items():

            labels_clean = detection['labels_clean']
            detections_clean = detection['detections_clean']
            labels_clean = labels_clean[:len(detections_clean)]

            logger.log(f"Results for {csm} CUSUM on {dataset} dataset:")
            fpr = sum(detections_clean) / len(detections_clean)
            logger.log(f"FPR: {fpr}")

            labels_fault = detection['labels_fault']
            detections_fault = detection['detections_fault']
            
            labels_fault = labels_fault[:len(detections_fault)]
            results = tpr_fpr_dd(labels_fault, detections_fault, simulation=simulation, sim_len=sim_length)

            logger.log(f"TPR: {results['TPR']}, DD: {results['DD'] if results['DD'] != -1 else 'NA'} \n")

    elif mode == 'robust':
        # Compute robustness metrics for each CUSUM method
        FARNs = []
        for csm, detection in cusum_detection_results.items():

            labels_clean = detection['labels_clean']
            detections_clean = detection['detections_clean']
            threshold = detection['threshold']

            FAR_N = robustness_metric(stat=detections_clean, 
                            threshold=threshold)
            
            FARNs.append(FAR_N)

        return FARNs

    elif mode == "resilience":
        # Compute resilience metrics for each CUSUM method
        for csm, detection in cusum_detection_results.items():
            
            cusum_stat_pos = detection['cusum_stat_pos']
            cusum_stat_neg = detection['cusum_stat_neg']

            labels_fault = detection['labels_fault']
            detections_fault = detection['detections_fault']
            threshold = detection['threshold']

            logger.log(f"Results for {csm} CUSUM on {dataset} dataset:")

            if simulation:
                labels_fault = np.array(labels_fault)
                padded = np.pad(labels_fault, (0, 1), mode='constant', constant_values=0)
                last_anomaly_in_data = np.where((padded[:-1] == 1) & (padded[1:] == 0))[0]
                last_anomaly_index = last_anomaly_in_data[0]
                last_anomaly_in_data = last_anomaly_index

                recovery_time, stability_time = resilience_metric(stat_pos=cusum_stat_pos,
                                                            stat_neg=cusum_stat_neg,
                                                            threshold=threshold,
                                                            last_anomaly_in_data=last_anomaly_in_data,
                                                            simulation=simulation,
                                                            sim_length=sim_length)

            logger.log(f"Recovery Time: {recovery_time}, Stability Time: {stability_time}")
            logger.log("")
            
        return recovery_time, stability_time
````

````python
# Function to run all CUSUMs for Time Measurement

def run_all_cusums_time_meas(
                        clean = None, 
                        fault = None,
                        data = None,
                        dataset = "TEP",
                        datasets_parameters = {},
                        cusums_parameters = {},
                        models_parameters = {},
                        run_on_residuals = False,
                        classic_threshold = False,
                        LENGTH = 0,
                        model = None):
    """
        Function to run all CUSUMs
        - Set the parameters in the cusums_parameters dictionary
        - Set the parameters in the datasets_parameters dictionary
        - Set the parameters in the models_parameters dictionary
        - Feel free to add remove datasets, CUSUMs and parameters
        Arguments:
            clean: clean data
            fault: fault data
            labels_fault: fault labels
            datasets_parameters: datasets parameters
            cusums_parameters: CUSUMs parameters
            run_on_residuals: True to run on residuals
            classic_threshold: True to use the classic threshold
        Returns:
            cusum_detection_results: CUSUMs detection results dict
                contains: labels, detections on clean and fault data
            dataset: dataset name
    """

    results_model_time = measure_model_time(
                            data = data,
                            datasets_parameters = datasets_parameters,
                            LENGTH = LENGTH,
                            model=model,
                            models_parameters = models_parameters,
                            dataset = dataset)
    # Params.
    datasets_parameters = datasets_parameters[dataset]
    
    param_estimation_size = datasets_parameters['param_estimation_size']
    variable = datasets_parameters['variable']
    simulation = datasets_parameters['simulation']
    sim_length = datasets_parameters['sim_length'] if datasets_parameters['simulation'] is True else 0
    

    param_size = int(len(clean) * param_estimation_size)
    param_estimation = np.array(clean[:param_size])
    clean = np.array(clean[param_size:])

    if LENGTH == 0:
        raise ValueError("LENGTH must be > 0 for time measurement")
    
    fault = fault[:LENGTH]
    #######################################################################################
    # Estimate Parameters for CUSUMs

    estimated_hs, estimated_ks, target_mean_std = \
                    estimate_cusums_parameters(param_estimation,
                                            classic_threshold = classic_threshold,
                                            cusums_parameters = cusums_parameters)
    
    target_mean = target_mean_std['all'][0]
    target_std = target_mean_std['all'][1]

    #######################################################################################
    # Detection Results    
    time_results = {}

    h_estimated_ho = estimated_hs['high_order']
    k_estimated_ho = estimated_ks['high_order']
    exp_x = target_mean_std['high_order']

    start = time.perf_counter()
    _, _, _, _ = high_order_cusum(fault, 
                                    window = cusums_parameters['high_order']['window'],
                                    alpha = cusums_parameters['high_order']['alpha'],
                                    exp_x=exp_x,
                                    k=k_estimated_ho, 
                                    h=h_estimated_ho,
                                    simulation=simulation,
                                    sim_length=sim_length
                                    )
    end = time.perf_counter()
    time_results['high_order'] = end - start

    h_estimated_classical = estimated_hs['classical']
    k_estimated_classical = estimated_ks['classical']

    start = time.perf_counter()
    _, _, _ = classical_cusum(fault, 
                                target_mean=target_mean, 
                                k=k_estimated_classical, 
                                h=h_estimated_classical,
                                simulation=simulation,
                                sim_length=sim_length)
    end = time.perf_counter()
    time_results['classical'] = end - start

    h_estimated_one = estimated_hs['one_cusum_scheme']
    k_estimated_one = estimated_ks['one_cusum_scheme']

    start = time.perf_counter()
    _, _ = one_cusum_scheme(fault, 
                            target_mean=target_mean, 
                            k=k_estimated_one, 
                            h=h_estimated_one, 
                            lamb=cusums_parameters['one_cusum_scheme']['lamb'],
                            simulation=simulation,
                            sim_length=sim_length) 
    end = time.perf_counter()
    time_results['one_cusum_scheme'] = end - start

    h_estimated_hampel = estimated_hs['hampel_acusum']
    k_estimated_hampel = estimated_ks['hampel_acusum']

    start = time.perf_counter()
    _, _, _ =  hampel_acusum(fault, 
                            a=cusums_parameters['hampel_acusum']['a'], 
                            b=cusums_parameters['hampel_acusum']['b'], 
                            r=cusums_parameters['hampel_acusum']['r'], 
                            gamma=cusums_parameters['hampel_acusum']['gamma'], 
                            h=h_estimated_hampel, 
                            k_plus=k_estimated_hampel, 
                            k_minus=k_estimated_hampel,
                            simulation=simulation,
                            sim_length=sim_length)
    end = time.perf_counter()
    time_results['hampel_acusum'] = end - start

    h_estimated_ucusum = estimated_hs['UCUSUM']

    start = time.perf_counter()
    _, _ = UCUSUM(fault,
                    target_mean=target_mean, 
                    h=h_estimated_ucusum,
                    simulation=simulation,
                    sim_length=sim_length)
    end = time.perf_counter()
    time_results['UCUSUM'] = end - start

    h_estimated_variance = estimated_hs['variance_cusum']
    k_estimated_variance = estimated_ks['variance_cusum']

    start = time.perf_counter()
    _, _, _ = variance_cusum(fault, 
                                    target_std=target_std, 
                                    window=cusums_parameters['variance_cusum']['window'], 
                                    k=k_estimated_variance, 
                                    h=h_estimated_variance,
                                    simulation=simulation,
                                    sim_length=sim_length)
    end = time.perf_counter()
    time_results['variance_cusum'] = end - start

    if run_on_residuals == False:
        results_model_time = None
    
    return time_results, results_model_time
````

````python
# Let us log the results, and save them to a file

class ExperimentLogger:
    def __init__(self, experiment_type, results_dir="results"):
        self.experiment_type = experiment_type
        os.makedirs(results_dir, exist_ok=True)
        timestamp = datetime.datetime.now().strftime("%Y-%m-%d_%H-%M-%S")
        self.filepath = os.path.join(results_dir, f"{experiment_type}_{timestamp}.txt")
        
        # Write header
        with open(self.filepath, "w") as f:
            f.write(f"Experiment: {experiment_type}\n")
            f.write(f"Started: {timestamp}\n")
            f.write("="*40 + "\n")

    def log(self, text):
        with open(self.filepath, "a", encoding="utf-8") as f:
            if isinstance(text, dict):
                text = {k: (v.to_dict() if isinstance(v, pd.Series) else v)
                        for k, v in text.items()}
                text = json.dumps(text)
            elif isinstance(text, pd.Series):
                text = json.dumps(text.to_dict())

            f.write(text + "\n")
````

````python
# Nors Computation
def compute_NORS(results_dict, cusumss, logger):

    for dataset_name, results in results_dict.items():
        levels = []
        far_n_values = []
        for entry in results:
            for tuple, far_n in entry.items():
                levels.append(tuple[2])  
                far_n_values.append(far_n)

                cusumss['high_order'].append(far_n[0])
                cusumss['classical'].append(far_n[1])
                cusumss['one_cusum'].append(far_n[2])
                cusumss['hampel_acusum'].append(far_n[3])
                cusumss['UCUSUM'].append(far_n[4])
                cusumss['variance_cusum'].append(far_n[5])

        for cusum, farns in cusumss.items():
            far_n_values = np.array(farns)
            far_n_values[1:] = (far_n_values[1:] - far_n_values[0])
            nors = 1 - (np.sum(far_n_values[1:]) / (len(far_n_values) - 1))
            #print("Dataset: {}, CUSUM: {}, NOISE NORS: {}".format(dataset_name, cusum, nors))
            logger.log(f"Dataset: {dataset_name}, CUSUM: {cusum}, NOISE NORS: {nors}\n")
````

````python
# DARTS Comparisson KMEANS ASCORERS WITH QUANTILE DETECTOR

from darts import TimeSeries
from darts.ad import QuantileDetector
from darts.ad.scorers import (
    KMeansScorer,
)

logger = ExperimentLogger("Comparissons")

DATASETS = ["TEP", "LBNL", "SKAB", "RMFD"]
PARAM_ESTIMATION_SIZE = 0.1
MAX_LENGTH = 0
WINDOW_SIZE = 30
K = 5
QUANTILE = 0.99

for dataset in DATASETS:

    clean_raw, fault_raw,labels_clean_raw,  labels_fault_raw = get_data_all(
                    data = data,
                    dataset = dataset,
                    MAX_LENGTH = MAX_LENGTH,
                    PARAM_ESTIMATION_SIZE = PARAM_ESTIMATION_SIZE,
                    datasets_parameters = datasets_parameters,
                    models_parameters = models_parameters,
                    residuals = False
                )

    variable = datasets_parameters[dataset]['variable']
    
    clean_raw = np.array(clean_raw[:, variable],dtype=float) 
    fault_raw = np.array(fault_raw[:, variable], dtype=float)

    logger.log("Running for {}".format(dataset))

    actual_length = len(clean_raw)
    start_after_training = int(actual_length * models_parameters['train_ratio'])

    # Create datasets
    trraining_raw = clean_raw[:start_after_training]
    clean_raw = clean_raw[start_after_training:]
    labels_clean_raw = labels_clean_raw[start_after_training:]

    # For fair comparison remove parameter estimation data
    param_size = int(len(clean_raw) * PARAM_ESTIMATION_SIZE)
    clean_raw = clean_raw[param_size:]
    labels_clean_raw = labels_clean_raw[param_size:]

    # Convert to TimeSeries
    series_train = TimeSeries.from_values(trraining_raw)
    series_clean_test = TimeSeries.from_values(clean_raw)
    series_anom_test = TimeSeries.from_values(fault_raw)


    # SCORER AND DETECTOR
    scorer = KMeansScorer(k=K, window=WINDOW_SIZE)
    scorer.fit(series_train)

    # Compute scores on training data for detector fitting
    train_scores = scorer.score(series_train)

    # Create Detector
    detector = QuantileDetector(high_quantile=QUANTILE)
    detector.fit(train_scores)

    # Compute scores
    anom_score_clean = scorer.score(series_clean_test)
    anom_score_anom = scorer.score(series_anom_test)

    # Compute detections
    binary_anom_clean = detector.detect(anom_score_clean).values().flatten()
    binary_anom_anom = detector.detect(anom_score_anom).values().flatten()

    # Compute FPR and TPR
    pad = np.zeros(WINDOW_SIZE - 1)
    binary_anom_clean = np.concatenate([pad, binary_anom_clean])
    binary_anom_anom = np.concatenate([pad, binary_anom_anom])

    FPR = np.sum(binary_anom_clean[WINDOW_SIZE-1:]) / len(binary_anom_clean[WINDOW_SIZE-1:])

    detection_results = tpr_fpr_dd(labels=labels_fault_raw, detections=binary_anom_anom)
    TPR = detection_results['TPR']
    DD = detection_results['DD']

    logger.log(
        "KMEANS SCORER RESULTS:\n"
        f"TPR: {TPR:.4f}, FPR: {FPR:.4f}, DD: {DD}\n"
    )
````

````python
# Connfig runner
def run_single_config(dataset, 
                      csm, 
                      WINDOW_SIZE,
                      clean_raw, fault_raw, 
                      clean_res, fault_res, 
                      labels_fault_raw, labels_fault_res,
                      logger = None):
    
    logger.log(f"Running for {dataset} with {csm}")

    try:       
        r1, r2, r3, r4, r5, r6 = test_optimize_cusum_parameters_oracle(
                dataset=dataset, 
                cusum=csm,  
                oracle=True,
                WINDOW_SIZE=WINDOW_SIZE,
                clean_raw = clean_raw, fault_raw = fault_raw, 
                clean_res = clean_res, fault_res = fault_res, 
                labels_fault_raw = labels_fault_raw, labels_fault_res = labels_fault_res
            )
        
        logger.log(f"Done for {dataset} with {csm}")

        return dataset, csm, r1, r2, r3, r4, r5, r6
    
    except Exception as e:
        print(f"[ERROR] Exception in {csm}: {e}")
        return dataset, csm, None, None, None, None, None, None



def test_optimize_cusum_parameters_oracle(dataset="TEP", 
                                   cusum = 'classical', 
                                   oracle = True,
                                   WINDOW_SIZE = 30,
                                   clean_raw = None, fault_raw = None, 
                                   clean_res = None, fault_res = None, 
                                   labels_fault_raw = None, 
                                   labels_fault_res = None):
    """ 
        Performance Analysis in terms of Parameter values

    """
    if any(x is None for x in [clean_raw, fault_raw, clean_res, fault_res, labels_fault_raw, labels_fault_res]):
        raise Exception("Missing data for experiment.")

    simulation = datasets_parameters[dataset]['simulation']

    if simulation:
        sim_len = datasets_parameters[dataset]['sim_length']
    else:
        sim_len = 0

    def run_opt(params):
        return optimize_cusum_parameters(
            data=params['data'],
            data_fault=params['data_fault'],
            labels=params['labels'],
            window_size=WINDOW_SIZE,
            cusum=cusum,
            oracle=oracle,
            sim_len=sim_len,
            classic_threshold=params['classic_threshold'],
            simulation=simulation,
            max_threshold=params['max_threshold']
        )

    # Let the Magic Happen
    tasks = [
        {
            'data': clean_raw,
            'data_fault': fault_raw,
            'labels': labels_fault_raw,
            'classic_threshold': True,
            'max_threshold': False
        },
        {
            'data': clean_raw,
            'data_fault': fault_raw,
            'labels': labels_fault_raw,
            'classic_threshold': False,
            'max_threshold': False
        },
        {
            'data': clean_res,
            'data_fault': fault_res,
            'labels': labels_fault_res,
            'classic_threshold': True,
            'max_threshold': False
        },
        {
            'data': clean_res,
            'data_fault': fault_res,
            'labels': labels_fault_res,
            'classic_threshold': False,
            'max_threshold': False
        },
        {
            'data': clean_raw,
            'data_fault': fault_raw,
            'labels': labels_fault_raw,
            'classic_threshold': False,
            'max_threshold': True
        },
        {
            'data': clean_res,
            'data_fault': fault_res,
            'labels': labels_fault_res,
            'classic_threshold': False,
            'max_threshold': True
        },
    ]

    with ThreadPoolExecutor(max_workers=24) as executor:
        results = list(executor.map(run_opt, tasks))

    if any(r is None for r in results):
        raise ValueError(f"[ERROR] One or more optimize_cusum_parameters results are None: {results}")

    return tuple(results)

def pareto_front(df, tpr_col="TPR", fpr_col="FPR"):
    return df[
        ~df.apply(
            lambda row: ((df[tpr_col] >= row[tpr_col]) & (df[fpr_col] <= row[fpr_col]) & ((df[tpr_col] > row[tpr_col]) | (df[fpr_col] < row[fpr_col]))).any(),
            axis=1
        )
    ]
````

````python
# Classical CUSUM Optimization

def optimize_classical_cusum(data = None, 
                             data_fault = None, 
                             labels = None, 
                             window_size = 30, 
                             oracle = True,
                             sim_len = None,
                             classic_threshold = False,
                             max_threshold = False,
                             simulation = False):
    """" 
    Optimize the parameters of the classical CUSUM
        - h, shift size.
    """

    results = pd.DataFrame(columns=['h', 'shift', 'FPR', 'TPR'])

    # Use X% of the data for parameter estimation
    param_est  = data[:int(len(data) * 0.1)]
    test = data[int(len(data) * 0.1):]
    
    target_mean = estimate_target_mean(param_est)
    target_std = np.std(param_est)

    rows, run_once = [], 0

    if classic_threshold:
        # Max-based Threshsold Optimization
        for shift_size in np.linspace(0.5, 5, 10):
            shift_size = round(shift_size, 1)
            run_once = 0
            for h in range(1, 7):

                if max_threshold and run_once == 0:
                    run_once = 1
                elif max_threshold and run_once == 1:
                    continue

                k_estimated_classical = estimate_k_from_shift(std_dev=target_std, 
                                shift_size=shift_size)
                
                h_estimated_classical = estimate_cusum_threshold(clean_data=param_est, 
                                        h= h,
                                        max_threshold = max_threshold)

                _, _, detections_clean = classical_cusum(test, 
                                            target_mean=target_mean, 
                                            k=k_estimated_classical, 
                                            h=h_estimated_classical,
                                            simulation=simulation,
                                            sim_length=sim_len)
                    
                FPR = sum(detections_clean) / len(detections_clean)

                if oracle:
                    if is_list_of_arrays(data_fault):
                        
                        TPRs = []
                        for v in range(len(data_fault)):
                            _, _, detections_fault = classical_cusum(data_fault[v], 
                                                target_mean=target_mean, 
                                                k=k_estimated_classical, 
                                                h=h_estimated_classical,
                                                simulation=simulation,
                                                sim_length=sim_len)
                            
                            TPRs.append(np.sum(detections_fault[labels == 1]) / np.sum(labels == 1))

                        TPR = np.mean(TPRs)
                        rows.append({'h': h, 'shift': shift_size, 'FPR': FPR, 'TPR': TPR})

                    else:
                        _, _, detections_fault = classical_cusum(data_fault, 
                                                target_mean=target_mean, 
                                                k=k_estimated_classical, 
                                                h=h_estimated_classical,
                                                simulation=simulation,
                                                sim_length=sim_len)
                        
                        TPR = np.sum(detections_fault[labels == 1]) / np.sum(labels == 1)
                    
                    rows.append({'h': h, 'shift': shift_size, 'FPR': FPR, 'TPR': TPR})
                else:
                    rows.append({'h': h, 'shift': shift_size, 'FPR': FPR})
    else:
        # Mean/std-based Threshold Optimization
        for shift_size in np.linspace(0.5, 5, 10):
            shift_size = round(shift_size, 1)
            run_once = 0
            for h in range(1, 7):

                if max_threshold and run_once == 0:
                    run_once = 1

                elif max_threshold and run_once == 1:
                    continue
            
                k_estimated_classical = estimate_k_from_shift(std_dev=target_std, 
                            shift_size=shift_size)

                pos_clean, neg_clean, _ = classical_cusum(param_est, 
                                        target_mean=target_mean, 
                                        k=k_estimated_classical, 
                                        h=0)

                pos_clean = np.maximum(abs(pos_clean), abs(neg_clean))

                h_estimated_classical = estimate_cusum_threshold(cusum_stats_pos=pos_clean,
                                                                 h = h,
                                                                 max_threshold=max_threshold)

                _, _, detections_clean = classical_cusum(test, 
                                        target_mean=target_mean, 
                                        k=k_estimated_classical, 
                                        h=h_estimated_classical,
                                        simulation=simulation,
                                        sim_length=sim_len)
                
                FPR = sum(detections_clean) / len(detections_clean)

                if oracle:
                    if is_list_of_arrays(data_fault):
                        TPRs = []
                        for v in range(len(data_fault)):
                            _, _, detections_fault = classical_cusum(data_fault[v], 
                                                target_mean=target_mean, 
                                                k=k_estimated_classical, 
                                                h=h_estimated_classical,
                                                simulation=simulation,
                                                sim_length=sim_len)
                            
                            TPRs.append(np.sum(detections_fault[labels == 1]) / np.sum(labels == 1))

                        TPR = np.mean(TPRs)

                        rows.append({'h': h, 'shift': shift_size, 'FPR': FPR, 'TPR': TPR})

                    else:
                        _, _, detections_fault = classical_cusum(data_fault, 
                                                target_mean=target_mean, 
                                                k=k_estimated_classical, 
                                                h=h_estimated_classical,
                                                simulation=simulation,
                                                sim_length=sim_len)
                        
                        TPR = np.sum(detections_fault[labels == 1]) / np.sum(labels == 1)

                        rows.append({'h': h, 'shift': shift_size, 'FPR': FPR, 'TPR': TPR})
                else:
                    rows.append({'h': h, 'shift': shift_size, 'FPR': FPR})

    results = pd.DataFrame(rows)
    return results
````

````python
# One-CUSUM CUSUM Optimization

def optimize_one_cusum(data = None, 
                             data_fault = None, 
                             labels = None, 
                             window_size = 30, 
                             oracle = False,
                             sim_len = None,
                             classic_threshold = False,
                             max_threshold = False,
                             simulation = False):
    """" 
    Optimize the parameters of the ONE CUSUM
        One CUSUM Scheme: h, k, lambda
    """

    results = pd.DataFrame(columns=['h', 'shift', 'lambda', 'FPR', 'TPR'])

    # Use X% of the data for parameter estimation
    param_est  = data[:int(len(data) * 0.1)]
    test = data[int(len(data) * 0.1):]
    
    target_mean = estimate_target_mean(param_est)
    target_std = np.std(param_est)

    rows, run_once = [], 0

    if classic_threshold:
        # Max-based Threshsold Optimization
        for Lambda in np.linspace(0.0, 1, 11):

            for shift_size in np.linspace(0.5, 5, 10):
                shift_size = round(shift_size, 1)
                run_once = 0
                for h in range(1, 7):

                    if max_threshold and run_once == 0:
                        run_once = 1
                    elif max_threshold and run_once == 1:
                        continue

                    k_estimated_one = estimate_k_from_shift(std_dev=target_std, 
                                    shift_size=shift_size)
                    
                    h_estimated_one = estimate_cusum_threshold(clean_data=param_est, 
                                            h= h,
                                            max_threshold = max_threshold)

                    _, detections_clean = one_cusum_scheme(test, 
                                          target_mean=target_mean, 
                                          k=k_estimated_one, 
                                          h=h_estimated_one, 
                                          lamb=Lambda, 
                                          simulation=simulation, 
                                          sim_length=sim_len) 
                        
                    FPR = sum(detections_clean) / len(detections_clean)

                    if oracle:
                        if is_list_of_arrays(data_fault):
                            TPRs = []
                            for v in range(len(data_fault)):
                                _, detections_fault = one_cusum_scheme(data_fault[v], 
                                          target_mean=target_mean, 
                                          k=k_estimated_one, 
                                          h=h_estimated_one, 
                                          lamb=Lambda, 
                                          simulation=simulation, 
                                          sim_length=sim_len) 
                            
                                TPRs.append(np.sum(detections_fault[labels == 1]) / np.sum(labels == 1))

                            TPR = np.mean(TPRs)

                            rows.append({'h': h, 'shift': shift_size, 'lambda': Lambda, 'FPR': FPR, 'TPR': TPR})

                        else:
                            _, detections_fault = one_cusum_scheme(data_fault, 
                                            target_mean=target_mean, 
                                            k=k_estimated_one, 
                                            h=h_estimated_one, 
                                            lamb=Lambda, 
                                            simulation=simulation, 
                                            sim_length=sim_len) 
                            
                            TPR = np.sum(detections_fault[labels == 1]) / np.sum(labels == 1)

                            rows.append({'h': h, 'shift': shift_size, 'lambda': Lambda, 'FPR': FPR, 'TPR': TPR})
                    else:
                        rows.append({'h': h, 'shift': shift_size, 'lambda': Lambda, 'FPR': FPR})
    else:
        # Mean/std-based Threshold Optimization
        for Lambda in np.linspace(0.0, 1, 11):

            for shift_size in np.linspace(0.5, 5, 10):
                shift_size = round(shift_size, 1)
                run_once = 0
                for h in range(1, 7):

                    if max_threshold and run_once == 0:
                        run_once = 1

                    elif max_threshold and run_once == 1:
                        continue
                
                    k_estimated_one = estimate_k_from_shift(std_dev=target_std, 
                                shift_size=shift_size)
                    
                    pos_clean, _ = one_cusum_scheme(param_est, 
                                          target_mean=target_mean, 
                                          k=k_estimated_one, 
                                          h=0, 
                                          lamb=Lambda) 


                    h_estimated_one = estimate_cusum_threshold(cusum_stats_pos=pos_clean,
                                                                    h = h,
                                                                    max_threshold=max_threshold)

                    _, detections_clean = one_cusum_scheme(test, 
                                          target_mean=target_mean, 
                                          k=k_estimated_one, 
                                          h=h_estimated_one, 
                                          lamb=Lambda,
                                          simulation=simulation, 
                                          sim_length=sim_len) 
                    
                    FPR = sum(detections_clean) / len(detections_clean)

                    if oracle:
                        if is_list_of_arrays(data_fault):
                            TPRs = []
                            for v in range(len(data_fault)):
                                _, detections_fault = one_cusum_scheme(data_fault[v], 
                                          target_mean=target_mean, 
                                          k=k_estimated_one, 
                                          h=h_estimated_one, 
                                          lamb=Lambda, 
                                          simulation=simulation, 
                                          sim_length=sim_len) 
                            
                                TPRs.append(np.sum(detections_fault[labels == 1]) / np.sum(labels == 1))

                            TPR = np.mean(TPRs)

                            rows.append({'h': h, 'shift': shift_size, 'lambda': Lambda, 'FPR': FPR, 'TPR': TPR})

                        else:
                            _, detections_fault = one_cusum_scheme(data_fault, 
                                            target_mean=target_mean, 
                                            k=k_estimated_one, 
                                            h=h_estimated_one, 
                                            lamb=Lambda, 
                                            simulation=simulation, 
                                            sim_length=sim_len) 
                            
                            TPR = np.sum(detections_fault[labels == 1]) / np.sum(labels == 1)

                            rows.append({'h': h, 'shift': shift_size, 'lambda': Lambda, 'FPR': FPR, 'TPR': TPR})
                    else:
                        rows.append({'h': h, 'shift': shift_size, 'lambda': Lambda, 'FPR': FPR})
                    

    results = pd.DataFrame(rows)
    return results
````

````python
# High Order CUSUM Optimization

def optimize_high_order_cusum(data = None, 
                             data_fault = None, 
                             labels = None, 
                             window_size = 30, 
                             oracle = False,
                             sim_len = None,
                             classic_threshold = False,
                             max_threshold = False,
                             simulation = False):
                             
    """" 
    Optimize the parameters of the High Order Cusum
        - h, alpha, window
    """
    results = pd.DataFrame(columns=['h', 'alpha', 'window', 'FPR', 'TPR'])

    # Use X% of the data for parameter estimation
    param_est  = data[:int(len(data) * 0.1)]
    test = data[int(len(data) * 0.1):]
    
    rows, run_once = [], 0

    if classic_threshold:
        # Max-based Threshsold Optimization
        for window in range(10, 60, 10):
            for alpha in range(0, 11):
                run_once = 0
                for h in range(1, 7):

                    if max_threshold and run_once == 0:
                        run_once = 1
                    elif max_threshold and run_once == 1:
                        continue

                    exp_x, k_estimated_ho = np.abs(estimate_k_from_stats(data = param_est, 
                                                         alpha=alpha,
                                                         window=window))
                    
                    h_estimated_ho = estimate_cusum_threshold(clean_data=param_est, 
                                                                h = h,
                                                                max_threshold = max_threshold)

                    _, _, _ , detections_clean = high_order_cusum_params(test, 
                                            window = window, 
                                            alpha = alpha,
                                            exp_x=exp_x,
                                            k=k_estimated_ho, 
                                            h=h_estimated_ho,
                                            simulation=simulation,
                                            sim_length=sim_len)
                        
                    FPR = sum(detections_clean) / len(detections_clean)

                    if oracle:
                        if is_list_of_arrays(data_fault):
                            TPRs = []
                            for v in range(len(data_fault)):
                                _, _, _ , detections_fault = high_order_cusum_params(data_fault[v], 
                                                        window = window, 
                                                        alpha = alpha,
                                                        exp_x=exp_x,
                                                        k=k_estimated_ho, 
                                                        h=h_estimated_ho,
                                                        simulation=simulation,
                                                        sim_length=sim_len)
                            
                                TPRs.append(np.sum(detections_fault[labels == 1]) / np.sum(labels == 1))

                            TPR = np.mean(TPRs)

                            rows.append({'h': h, 'alpha': alpha, 'window': window, 'FPR': FPR, 'TPR': TPR})

                        else:
                            _, _, _ , detections_fault = high_order_cusum_params(data_fault, 
                                                            window = window, 
                                                            alpha = alpha,
                                                            exp_x=exp_x,
                                                            k=k_estimated_ho, 
                                                            h=h_estimated_ho,
                                                            simulation=simulation,
                                                            sim_length=sim_len)
                            
                            TPR = np.sum(detections_fault[labels == 1]) / np.sum(labels == 1)

                            rows.append({'h': h, 'alpha': alpha, 'window': window, 'FPR': FPR, 'TPR': TPR})
                    else:
                        rows.append({'h': h, 'alpha': alpha, 'window': window, 'FPR': FPR})
    else:
        for window in range(10, 60, 10):
            
            # Mean/std-based Threshold Optimization
            for alpha in range(0, 11):
                run_once = 0
                
                for h in range(1, 7):

                    if max_threshold and run_once == 0:
                        run_once = 1

                    elif max_threshold and run_once == 1:
                        continue
                
                    exp_x, k_estimated_ho = np.abs(estimate_k_from_stats(data = param_est, 
                                                         alpha=alpha,
                                                         window=window))

                    pos_clean, neg_clean, _, _ = high_order_cusum_params(param_est, 
                                            window = window, 
                                            alpha = alpha,
                                            exp_x=exp_x,
                                            k=k_estimated_ho)
                    
                    pos_clean = np.maximum(abs(pos_clean), abs(neg_clean))

                    h_estimated_ho = estimate_cusum_threshold(cusum_stats_pos=pos_clean, 
                                            cusum_stats_neg=neg_clean,
                                            clean_data=param_est, 
                                            h=h, 
                                            csm_type="high_order",
                                            window=window,
                                            alpha=alpha,
                                            max_threshold=max_threshold)
                    
                    _, _, _ , detections_clean = high_order_cusum(test, 
                                            window = window, 
                                            alpha = alpha,
                                            exp_x=exp_x,
                                            k=k_estimated_ho, 
                                            h=h_estimated_ho,
                                            simulation=simulation,
                                            sim_length=sim_len)
                    
                    FPR = sum(detections_clean) / len(detections_clean)

                    if oracle:
                        if is_list_of_arrays(data_fault):
                            TPRs = []
                            for v in range(len(data_fault)):
                                _, _, _ , detections_fault = high_order_cusum_params(data_fault[v], 
                                                        window = window, 
                                                        alpha = alpha,
                                                        exp_x=exp_x,
                                                        k=k_estimated_ho, 
                                                        h=h_estimated_ho,
                                                        simulation=simulation,
                                                        sim_length=sim_len)
                            
                                TPRs.append(np.sum(detections_fault[labels == 1]) / np.sum(labels == 1))

                            TPR = np.mean(TPRs)

                            rows.append({'h': h, 'alpha': alpha, 'window': window, 'FPR': FPR, 'TPR': TPR})

                        else:

                            _, _, _ , detections_fault = high_order_cusum_params(data_fault, 
                                                            window = window, 
                                                            alpha = alpha,
                                                            exp_x=exp_x,
                                                            k=k_estimated_ho, 
                                                            h=h_estimated_ho,
                                                            simulation=simulation,
                                                            sim_length=sim_len)
                                
                            TPR = np.sum(detections_fault[labels == 1]) / np.sum(labels == 1)

                            rows.append({'h': h, 'alpha': alpha, 'window': window, 'FPR': FPR, 'TPR': TPR})
                    else:
                        rows.append({'h': h, 'alpha': alpha, 'window': window, 'FPR': FPR})

    results = pd.DataFrame(rows)
    return results
                    
````

````python
# Run over all CUSUMs and parameters
# There will be to much redundant code, for all the cusum functions....
# I will fix this later, or just leave it like this, it works.

def optimize_cusum_parameters(data = None, 
                              data_fault = None, 
                              labels = None, 
                              window_size = 30, 
                              cusum = 'classical',
                              oracle = True,
                              sim_len = None,
                              classic_threshold = False,
                              max_threshold = False,
                              simulation = False):
    if data is None:
        raise ValueError("Data must be provided")
    if labels is None:
        raise ValueError("Labels must be provided")
    if oracle is True and data_fault is None:
        raise ValueError("Data fault must be provided for oracle mode")

    if cusum == 'classical':
        return optimize_classical_cusum(data = data, 
                                        data_fault = data_fault,
                                        labels = labels, 
                                        window_size = window_size,
                                        oracle = oracle,
                                        sim_len = sim_len,
                                        classic_threshold = classic_threshold,
                                        max_threshold = max_threshold,
                                        simulation = simulation)
    elif cusum == 'one_cusum_scheme':
        return optimize_one_cusum(data = data, 
                                    data_fault = data_fault,
                                    labels = labels, 
                                    window_size = window_size,
                                    oracle = oracle,
                                    sim_len = sim_len,
                                    classic_threshold = classic_threshold,
                                    max_threshold = max_threshold,
                                    simulation = simulation)

    elif cusum == 'hampel_acusum':
        return optimize_hampel_acusum(data = data, 
                                    data_fault = data_fault,
                                    labels = labels, 
                                    window_size = window_size,
                                    oracle = oracle,
                                    sim_len = sim_len,
                                    classic_threshold = classic_threshold,
                                    max_threshold = max_threshold,
                                    simulation = simulation)
    elif cusum == 'UCUSUM':
        return optimize_ucusum(data = data, 
                                    data_fault = data_fault,
                                    labels = labels, 
                                    window_size = window_size,
                                    oracle = oracle,
                                    sim_len = sim_len,
                                    classic_threshold = classic_threshold,
                                    max_threshold = max_threshold,
                                    simulation = simulation)
    elif cusum == 'variance_cusum':
        return optimize_variance_cusum(data = data, 
                                    data_fault = data_fault,
                                    labels = labels, 
                                    window_size = window_size,
                                    oracle = oracle,
                                    sim_len = sim_len,
                                    classic_threshold = classic_threshold,
                                    max_threshold = max_threshold,
                                    simulation = simulation)

    elif cusum == 'high_order':
        return optimize_high_order_cusum(data = data, 
                                    data_fault = data_fault,
                                    labels = labels, 
                                    window_size = window_size,
                                    oracle = oracle,
                                    sim_len = sim_len,
                                    classic_threshold = classic_threshold,
                                    max_threshold = max_threshold,
                                    simulation = simulation)
    
    else:
        raise ValueError(f"Unknown CUSUM type: {cusum}")
````
